```python
# Reviewed numeric snapshots from the immutable version 1 run.
# Expand this cell to inspect inputs; no source-code ZIP, model weights or training are loaded.
import hashlib, json
evidence_json='{"summary":{"input_results_sha256":"718355cfdc205bb7652a79d57e517231c8b155baaf01de3397a6b3a5693ca8c6","status":"completed","attempted_runs":120,"failed_runs":[],"incomplete_confirmation_blocks":[],"skipped":["Equal-budget boundary extension: insufficient time","Optional primary block 104: insufficient reserved time","Optional primary block 105: insufficient reserved time"],"validation":{"wildlife":{"metric":"val_map50_95","units":"AP points (100 times mAP)","joint_blocks":[101,102,103],"interval_method":"Student t interval across independent paired joint training/subset/corruption blocks; approximate at small n; no multiplicity adjustment","conditions":[{"optimizer":"AdamW","fraction":0.2,"missing_probability":0.0,"n_joint_blocks":3,"mean":77.16149583578266,"sd":1.386617757407881,"ci95":[73.71694637286576,80.60604529869956],"values":[77.99847787534947,77.92508121126345,75.56092842073505]},{"optimizer":"MuSGD","fraction":0.2,"missing_probability":0.0,"n_joint_blocks":3,"mean":76.45201628342387,"sd":0.4296945048000814,"ci95":[75.38459595953134,77.51943660731641],"values":[76.89810727304301,76.04084984562589,76.41709173160274]},{"optimizer":"SGD","fraction":0.2,"missing_probability":0.0,"n_joint_blocks":3,"mean":77.88202457327709,"sd":0.1882207193354869,"ci95":[77.41445838624882,78.34959076030536],"values":[77.66527689083249,78.0042681947189,77.97652863427989]},{"optimizer":"AdamW","fraction":0.2,"missing_probability":0.2,"n_joint_blocks":3,"mean":72.92866444827824,"sd":0.6651481307613413,"ci95":[71.27634489277423,74.58098400378225],"values":[73.44201047074046,72.17724005324717,73.16674282084709]},{"optimizer":"MuSGD","fraction":0.2,"missing_probability":0.2,"n_joint_blocks":3,"mean":71.56146565267052,"sd":0.20142737045036715,"ci95":[71.06109232556226,72.06183897977878],"values":[71.33315517845486,71.63716661707541,71.71407516248131]},{"optimizer":"SGD","fraction":0.2,"missing_probability":0.2,"n_joint_blocks":3,"mean":72.93530370421217,"sd":0.9113140957449326,"ci95":[70.67147399175059,75.19913341667376],"values":[73.96286171193803,72.61795989748116,72.22508950321733]},{"optimizer":"AdamW","fraction":1.0,"missing_probability":0.0,"n_joint_blocks":3,"mean":82.17500832636675,"sd":0.47026008931852414,"ci95":[81.00681750417401,83.34319914855949],"values":[82.49851882530577,82.39094378291611,81.63556237087836]},{"optimizer":"MuSGD","fraction":1.0,"missing_probability":0.0,"n_joint_blocks":3,"mean":82.00030787178346,"sd":0.22039010450970412,"ci95":[81.4528285018811,82.54778724168582],"values":[81.92917754111913,82.24747916937669,81.82426690485457]},{"optimizer":"SGD","fraction":1.0,"missing_probability":0.0,"n_joint_blocks":3,"mean":82.8391431542353,"sd":1.1723165954050472,"ci95":[79.92694728951496,85.75133901895565],"values":[82.94324319084485,83.95593811566286,81.61824815619818]},{"optimizer":"AdamW","fraction":1.0,"missing_probability":0.2,"n_joint_blocks":3,"mean":79.27793334818742,"sd":0.603436757018327,"ci95":[77.77891334344045,80.77695335293438],"values":[79.91070828730554,78.70889428643302,79.21419747082368]},{"optimizer":"MuSGD","fraction":1.0,"missing_probability":0.2,"n_joint_blocks":3,"mean":79.89605614621063,"sd":0.4599875894611068,"ci95":[78.75338362830732,81.03872866411393],"values":[80.2824503085391,79.38724340328814,80.01847472680464]},{"optimizer":"SGD","fraction":1.0,"missing_probability":0.2,"n_joint_blocks":3,"mean":80.26657159763963,"sd":0.7242835907507483,"ci95":[78.46735141587605,82.0657917794032],"values":[80.93517206366853,80.36737684890485,79.4971658803455]}],"paired_optimizer_contrasts":[{"contrast":"MuSGD minus AdamW","fraction":0.2,"missing_probability":0.0,"n_joint_blocks":3,"mean":-0.709479552358779,"sd":1.4113956232932992,"ci95":[-4.215580646337577,2.796621541620019],"values":[-1.1003706023064552,-1.8842313656375675,0.8561633108676858]},{"contrast":"SGD minus AdamW","fraction":0.2,"missing_probability":0.0,"n_joint_blocks":3,"mean":0.7205287374944414,"sd":1.4823853887305578,"ci95":[-2.961920709953176,4.402978184942059],"values":[-0.3332009845169681,0.07918698345545927,2.415600213544833]},{"contrast":"MuSGD minus AdamW","fraction":0.2,"missing_probability":0.2,"n_joint_blocks":3,"mean":-1.3671987956077158,"sd":0.787875512307534,"ci95":[-3.324390067871218,0.5899924766557865],"values":[-2.108855292285594,-0.5400734361717618,-1.4526676583657916]},{"contrast":"SGD minus AdamW","fraction":0.2,"missing_probability":0.2,"n_joint_blocks":3,"mean":0.006639255933930481,"sd":0.8222222108419818,"ci95":[-2.0358739454320562,2.049152457299917],"values":[0.520851241197573,0.44071984423398325,-0.9416533176297648]},{"contrast":"MuSGD minus AdamW","fraction":1.0,"missing_probability":0.0,"n_joint_blocks":3,"mean":-0.17470045458328384,"sd":0.37998700495258564,"ci95":[-1.1186405035493636,0.769239594382796],"values":[-0.5693412841866419,-0.14346461353942352,0.1887045339762139]},{"contrast":"SGD minus AdamW","fraction":1.0,"missing_probability":0.0,"n_joint_blocks":3,"mean":0.6641348278685529,"sd":0.8136527502871232,"ci95":[-1.3570906533640166,2.6853603091011227],"values":[0.44472436553908423,1.5649943327467497,-0.017314214680175244]},{"contrast":"MuSGD minus AdamW","fraction":1.0,"missing_probability":0.2,"n_joint_blocks":3,"mean":0.6181227980232166,"sd":0.2224681762287064,"ci95":[0.06548121179601718,1.170764384250416],"values":[0.3717420212335676,0.6783491168551259,0.8042772559809563]},{"contrast":"SGD minus AdamW","fraction":1.0,"missing_probability":0.2,"n_joint_blocks":3,"mean":0.9886382494522191,"sd":0.688456532778416,"ci95":[-0.7215825865023276,2.698859085406766],"values":[1.0244637763629938,1.6584825624718413,0.2829684095218221]}],"three_way_interactions":[{"contrast":"MuSGD minus AdamW: scarcity x missing-box interaction","n_joint_blocks":3,"mean":-1.4505424958554374,"sd":1.776735093291764,"ci95":[-5.864197144837054,2.96311215312618],"values":[-1.9495679953993483,0.5223441990712563,-2.92440369123822]},{"contrast":"SGD minus AdamW: scarcity x missing-box interaction","n_joint_blocks":3,"mean":-1.038392903144177,"sd":2.2682467578105583,"ci95":[-6.673030213706976,4.596244407418622],"values":[0.27431281489063153,0.26804463105343235,-3.657536155376595]}]},"voc2007":{"metric":"val_map50_95","units":"AP points (100 times mAP)","joint_blocks":[101,102,103],"interval_method":"Student t interval across independent paired joint training/subset/corruption blocks; approximate at small n; no multiplicity adjustment","conditions":[{"optimizer":"AdamW","fraction":0.2,"missing_probability":0.0,"n_joint_blocks":3,"mean":47.74337917701948,"sd":0.09624518437333424,"ci95":[47.504292884946274,47.98246546909268],"values":[47.76468094996794,47.638267656591225,47.827188924499275]},{"optimizer":"MuSGD","fraction":0.2,"missing_probability":0.0,"n_joint_blocks":3,"mean":49.52854240739748,"sd":0.6541664856968906,"ci95":[47.90350277053479,51.15358204426017],"values":[49.81167078270657,49.99345372075691,48.78050271872897]},{"optimizer":"AdamW","fraction":0.2,"missing_probability":0.2,"n_joint_blocks":3,"mean":43.743325730375865,"sd":0.4111986932904111,"ci95":[42.72185154936336,44.76479991138837],"values":[43.30643073896972,43.80075633185069,44.12279012030719]},{"optimizer":"MuSGD","fraction":0.2,"missing_probability":0.2,"n_joint_blocks":3,"mean":45.22410963012578,"sd":0.21118855146676002,"ci95":[44.69948818514378,45.748731075107784],"values":[45.13910459749563,45.46455459411514,45.06866969876656]},{"optimizer":"AdamW","fraction":1.0,"missing_probability":0.0,"n_joint_blocks":3,"mean":54.88750378940945,"sd":0.6662278822352445,"ci95":[53.232501982549834,56.542505596269066],"values":[55.28173070993662,55.262490357476935,54.1182903008148]},{"optimizer":"MuSGD","fraction":1.0,"missing_probability":0.0,"n_joint_blocks":3,"mean":55.63188703185454,"sd":0.223275526688147,"ci95":[55.07723987590447,56.18653418780461],"values":[55.87531652361157,55.43662962675667,55.58371494519538]},{"optimizer":"AdamW","fraction":1.0,"missing_probability":0.2,"n_joint_blocks":3,"mean":52.98585588628073,"sd":0.2987889422879875,"ci95":[52.243623006898346,53.72808876566312],"values":[53.108653329966145,52.645234273360266,53.203680055515775]},{"optimizer":"MuSGD","fraction":1.0,"missing_probability":0.2,"n_joint_blocks":3,"mean":54.62464854094018,"sd":0.7980694417992218,"ci95":[52.64213414399576,56.607162937884596],"values":[54.84946277803536,55.286197979980415,53.738284864804754]}],"paired_optimizer_contrasts":[{"contrast":"MuSGD minus AdamW","fraction":0.2,"missing_probability":0.0,"n_joint_blocks":3,"mean":1.7851632303780007,"sd":0.7366996284658033,"ci95":[-0.044900098903679186,3.6152265596596806],"values":[2.0469898327386273,2.355186064165682,0.9533137942296932]},{"contrast":"MuSGD minus AdamW","fraction":0.2,"missing_probability":0.2,"n_joint_blocks":3,"mean":1.48078389974991,"sd":0.47087335500602745,"ci95":[0.3110696411355318,2.6504981583642886],"values":[1.8326738585259073,1.6637982622644487,0.9458795784593743]},{"contrast":"MuSGD minus AdamW","fraction":1.0,"missing_probability":0.0,"n_joint_blocks":3,"mean":0.7443832424450939,"sd":0.6587179809145168,"ci95":[-0.8919629353324136,2.3807294202226013],"values":[0.5935858136749572,0.17413926927973833,1.4654246443805863]},{"contrast":"MuSGD minus AdamW","fraction":1.0,"missing_probability":0.2,"n_joint_blocks":3,"mean":1.6387926546594451,"sd":1.0568786688616425,"ci95":[-0.9866395033717146,4.264224812690605],"values":[1.7408094480692138,2.6409637066201475,0.5346048092889744]}],"three_way_interactions":[{"contrast":"MuSGD minus AdamW: scarcity x missing-box interaction","n_joint_blocks":3,"mean":-1.198788742842442,"sd":2.0456603219954954,"ci95":[-6.280490694079802,3.8829132083949176],"values":[-1.3615396086069764,-3.1582122392416423,0.923385619321293]}]}},"frozen_test":{"wildlife":{"metric":"test_map50_95","units":"AP points (100 times mAP)","joint_blocks":[101,102,103],"interval_method":"Student t interval across independent paired joint training/subset/corruption blocks; approximate at small n; no multiplicity adjustment","conditions":[{"optimizer":"AdamW","fraction":0.2,"missing_probability":0.0,"n_joint_blocks":3,"mean":80.29947577580687,"sd":0.34497127036422065,"ci95":[79.44251963363531,81.15643191797842],"values":[80.54489559055409,79.90504539943288,80.44848633743362]},{"optimizer":"MuSGD","fraction":0.2,"missing_probability":0.0,"n_joint_blocks":3,"mean":81.08452218056165,"sd":0.8976259612694601,"ci95":[78.85469567915368,83.31434868196962],"values":[81.94562328945827,80.15436075099383,81.15358250123286]},{"optimizer":"SGD","fraction":0.2,"missing_probability":0.0,"n_joint_blocks":3,"mean":81.1696383600305,"sd":0.959791566174579,"ci95":[78.78538393510587,83.55389278495514],"values":[81.71816992810189,80.06138535209884,81.72935979989079]},{"optimizer":"AdamW","fraction":0.2,"missing_probability":0.2,"n_joint_blocks":3,"mean":74.86457082521181,"sd":0.979026674732429,"ci95":[72.4325337417296,77.29660790869403],"values":[74.83585436748055,73.9002182931122,75.85763981504269]},{"optimizer":"MuSGD","fraction":0.2,"missing_probability":0.2,"n_joint_blocks":3,"mean":75.12339271437406,"sd":0.7715779111171595,"ci95":[73.20668692783813,77.04009850090999],"values":[74.90712790199352,74.48302353665257,75.9800267044761]},{"optimizer":"SGD","fraction":0.2,"missing_probability":0.2,"n_joint_blocks":3,"mean":77.59442339380199,"sd":1.0230511849770723,"ci95":[75.05302336418107,80.1358234234229],"values":[78.48689191115798,76.47792151228272,77.81845675796525]},{"optimizer":"AdamW","fraction":1.0,"missing_probability":0.0,"n_joint_blocks":3,"mean":84.31724622109735,"sd":0.9944757224485091,"ci95":[81.84683157557346,86.78766086662124],"values":[84.73258556376398,85.03672372544028,83.1824293740878]},{"optimizer":"MuSGD","fraction":1.0,"missing_probability":0.0,"n_joint_blocks":3,"mean":85.06198287665163,"sd":0.210401848153566,"ci95":[84.53931571103786,85.5846500422654],"values":[84.82554962459747,85.2286084450216,85.1317905603358]},{"optimizer":"SGD","fraction":1.0,"missing_probability":0.0,"n_joint_blocks":3,"mean":84.59368179035432,"sd":0.17870081160322487,"ci95":[84.14976436513587,85.03759921557278],"values":[84.62908824542659,84.39992809484932,84.75202903078709]},{"optimizer":"AdamW","fraction":1.0,"missing_probability":0.2,"n_joint_blocks":3,"mean":82.78273963752798,"sd":0.7798248365127248,"ci95":[80.84554735261119,84.71993192244477],"values":[82.09520727940415,83.63009415843432,82.62291747474549]},{"optimizer":"MuSGD","fraction":1.0,"missing_probability":0.2,"n_joint_blocks":3,"mean":82.5316042508488,"sd":1.0325400486094896,"ci95":[79.96663257723723,85.09657592446037],"values":[83.46094692039952,81.42010944409925,82.71375638804763]},{"optimizer":"SGD","fraction":1.0,"missing_probability":0.2,"n_joint_blocks":3,"mean":82.85987641201662,"sd":1.2787310287365932,"ci95":[79.68333244038612,86.03642038364711],"values":[83.98426168772343,81.46884642118977,83.12652112713667]}],"paired_optimizer_contrasts":[{"contrast":"MuSGD minus AdamW","fraction":0.2,"missing_probability":0.0,"n_joint_blocks":3,"mean":0.785046404754793,"sd":0.5798548345708977,"ci95":[-0.6553928571256725,2.2254856666352585],"values":[1.4007276989041806,0.24931535156096363,0.7050961637992348]},{"contrast":"SGD minus AdamW","fraction":0.2,"missing_probability":0.0,"n_joint_blocks":3,"mean":0.8701625842236438,"sd":0.6205251444649644,"ci95":[-0.6713073282119935,2.411632496659281],"values":[1.1732743375478027,0.15633995266596257,1.280873462457166]},{"contrast":"MuSGD minus AdamW","fraction":0.2,"missing_probability":0.2,"n_joint_blocks":3,"mean":0.2588218891622492,"sd":0.28173933732863177,"ci95":[-0.4410574235706793,0.9587012018951777],"values":[0.07127353451297225,0.582805243540363,0.12238688943341236]},{"contrast":"SGD minus AdamW","fraction":0.2,"missing_probability":0.2,"n_joint_blocks":3,"mean":2.7298525685901707,"sd":0.8553206931909593,"ci95":[0.6051181790204057,4.854586958159936],"values":[3.651037543677438,2.577703219170513,1.9608169429225608]},{"contrast":"MuSGD minus AdamW","fraction":1.0,"missing_probability":0.0,"n_joint_blocks":3,"mean":0.7447366555542789,"sd":1.0444072573204366,"ci95":[-1.8497147987489972,3.339188109857555],"values":[0.09296406083350028,0.19188471958132558,1.949361186248011]},{"contrast":"SGD minus AdamW","fraction":1.0,"missing_probability":0.0,"n_joint_blocks":3,"mean":0.2764355692569757,"sd":1.1512196967785957,"ci95":[-2.583352693985075,3.136223832499026],"values":[-0.10349731833738662,-0.6367956305909694,1.5695996566992831]},{"contrast":"MuSGD minus AdamW","fraction":1.0,"missing_probability":0.2,"n_joint_blocks":3,"mean":-0.2511353866791873,"sd":1.8122254227086652,"ci95":[-4.752952901366677,4.250682128008303],"values":[1.365739640995367,-2.209984714335067,0.09083891330213811]},{"contrast":"SGD minus AdamW","fraction":1.0,"missing_probability":0.2,"n_joint_blocks":3,"mean":0.07713677448863976,"sd":2.0585534644625403,"ci95":[-5.036593518173679,5.1908670671509585],"values":[1.8890544083192906,-2.1612477372445538,0.5036036523911824]}],"three_way_interactions":[{"contrast":"MuSGD minus AdamW: scarcity x missing-box interaction","n_joint_blocks":3,"mean":0.4696475266409224,"sd":2.7586031514797353,"ci95":[-6.383102593618275,7.322397646900119],"values":[-2.602229744553075,2.735359325895792,1.2758129985800504]},{"contrast":"SGD minus AdamW: scarcity x missing-box interaction","n_joint_blocks":3,"mean":2.0589887791348627,"sd":1.751412208125878,"ci95":[-2.291760335836633,6.4097378941063585],"values":[0.48521147947295784,3.9458153731581347,1.7459394847734955]}]},"voc2007":{"metric":"test_map50_95","units":"AP points (100 times mAP)","joint_blocks":[101,102,103],"interval_method":"Student t interval across independent paired joint training/subset/corruption blocks; approximate at small n; no multiplicity adjustment","conditions":[{"optimizer":"AdamW","fraction":0.2,"missing_probability":0.0,"n_joint_blocks":3,"mean":46.59055479704567,"sd":0.20262181735083443,"ci95":[46.08721429934731,47.09389529474403],"values":[46.364377780233845,46.65179189596125,46.755494714941904]},{"optimizer":"MuSGD","fraction":0.2,"missing_probability":0.0,"n_joint_blocks":3,"mean":48.14443844153133,"sd":0.5739828919226496,"ci95":[46.71858589382443,49.570290989238224],"values":[48.657174140409666,47.52436337805634,48.251777806127976]},{"optimizer":"AdamW","fraction":0.2,"missing_probability":0.2,"n_joint_blocks":3,"mean":42.18100617142262,"sd":0.604698532943073,"ci95":[40.67885174151724,43.68316060132799],"values":[41.50315063810411,42.374867668033936,42.6650002081298]},{"optimizer":"MuSGD","fraction":0.2,"missing_probability":0.2,"n_joint_blocks":3,"mean":43.85217574483554,"sd":0.6696226921810625,"ci95":[42.188740762565395,45.51561072710569],"values":[43.460094497666994,43.471069271400445,44.62536346543918]},{"optimizer":"AdamW","fraction":1.0,"missing_probability":0.0,"n_joint_blocks":3,"mean":54.15226717721152,"sd":0.3191984502044687,"ci95":[53.35933426953616,54.94520008488688],"values":[53.83255670376603,54.15329369962265,54.47095112824588]},{"optimizer":"MuSGD","fraction":1.0,"missing_probability":0.0,"n_joint_blocks":3,"mean":55.25414493664988,"sd":0.15676262255709375,"ci95":[54.864724994167744,55.643564879132015],"values":[55.07799175754122,55.30613741684272,55.3783056355657]},{"optimizer":"AdamW","fraction":1.0,"missing_probability":0.2,"n_joint_blocks":3,"mean":52.240049180798486,"sd":0.0496349049495057,"ci95":[52.116749241595805,52.36334912000117],"values":[52.220395394217846,52.20325075592229,52.296501392255315]},{"optimizer":"MuSGD","fraction":1.0,"missing_probability":0.2,"n_joint_blocks":3,"mean":52.97091296632983,"sd":0.5319450149092477,"ci95":[51.64948829423255,54.29233763842711],"values":[52.58796136662988,53.57829326615815,52.74648426620146]}],"paired_optimizer_contrasts":[{"contrast":"MuSGD minus AdamW","fraction":0.2,"missing_probability":0.0,"n_joint_blocks":3,"mean":1.5538836444856614,"sd":0.7118623770681863,"ci95":[-0.2144805319437395,3.322247820915062],"values":[2.2927963601758194,0.8725714820950914,1.4962830911860736]},{"contrast":"MuSGD minus AdamW","fraction":0.2,"missing_probability":0.2,"n_joint_blocks":3,"mean":1.6711695734129266,"sd":0.4979398035954808,"ci95":[0.434218529135169,2.908120617690684],"values":[1.9569438595628874,1.0962016033665145,1.9603632573093777]},{"contrast":"MuSGD minus AdamW","fraction":1.0,"missing_probability":0.0,"n_joint_blocks":3,"mean":1.1018777594383595,"sd":0.17470764325679805,"ci95":[0.6678799142985025,1.5358756045782167],"values":[1.2454350537751901,1.1528437172200645,0.9073545073198241]},{"contrast":"MuSGD minus AdamW","fraction":1.0,"missing_probability":0.2,"n_joint_blocks":3,"mean":0.7308637855313475,"sd":0.559395037907978,"ci95":[-0.6587505238846453,2.12047809494734],"values":[0.3675659724120317,1.3750425102358577,0.4499828739461531]}],"three_way_interactions":[{"contrast":"MuSGD minus AdamW: scarcity x missing-box interaction","n_joint_blocks":3,"mean":0.48829990283427716,"sd":0.4623564944829163,"ci95":[-0.6602573013691855,1.6368571070377398],"values":[0.5420165807502264,0.0014313282556299711,0.9214517994969751]}]}},"test_status":"completed","limitations":["Short-budget fine-tuning recipes, not isolated orthogonalization effects or convergence results.","Joint blocks vary training, subset and annotation deletion seeds together; intervals do not isolate these variance sources.","Approximate t intervals at n=3 are unstable. Cell contrasts are exploratory; no multiple-comparison claims.","One checkpoint lineage, uncertain pretrained-image overlap, and unresolved source-image rights limit generalization.","VOC-derived all-object mAP50-95 is not official VOC AP; validation uses a frozen 512-image sample.","Hyperparameters selected at 10 exposure blocks and transferred to 20/60-block schedules; tuning is not convergence-based."],"incomplete_test_blocks":{"wildlife":[],"voc2007":[]}},"selection":{"selected":{"AdamW":{"optimizer":"AdamW","lr0":0.0003,"momentum":0.9,"weight_decay":0.0005},"MuSGD":{"optimizer":"MuSGD","lr0":0.003,"momentum":0.9,"weight_decay":0.0005},"SGD":{"optimizer":"SGD","lr0":0.003,"momentum":0.9,"weight_decay":0.0}},"diagnostics":{"AdamW":{"ranked":[{"candidate_id":"eb609f942ee0","hyperparameters":{"optimizer":"AdamW","lr0":0.0003,"momentum":0.9,"weight_decay":0.0005},"mean_anchor_val_map":0.7778696404142218},{"candidate_id":"7555534c5a8e","hyperparameters":{"optimizer":"AdamW","lr0":0.0003,"momentum":0.9,"weight_decay":0.0},"mean_anchor_val_map":0.7661279641494985},{"candidate_id":"1543a592cd37","hyperparameters":{"optimizer":"AdamW","lr0":0.001,"momentum":0.9,"weight_decay":0.0},"mean_anchor_val_map":0.7156664712824967},{"candidate_id":"a54004a2c417","hyperparameters":{"optimizer":"AdamW","lr0":0.001,"momentum":0.9,"weight_decay":0.0005},"mean_anchor_val_map":0.7137647210968419},{"candidate_id":"84fae146d963","hyperparameters":{"optimizer":"AdamW","lr0":0.003,"momentum":0.9,"weight_decay":0.0005},"mean_anchor_val_map":0.45856027398738025},{"candidate_id":"3a07159f60a4","hyperparameters":{"optimizer":"AdamW","lr0":0.003,"momentum":0.9,"weight_decay":0.0},"mean_anchor_val_map":0.4380522295404367}],"selection_margin_ap_points":1.1741676264723333,"tried_candidates":6,"valid_candidates":6},"MuSGD":{"ranked":[{"candidate_id":"d4ffe71ced46","hyperparameters":{"optimizer":"MuSGD","lr0":0.003,"momentum":0.9,"weight_decay":0.0005},"mean_anchor_val_map":0.7788370106059557},{"candidate_id":"eddb709374b2","hyperparameters":{"optimizer":"MuSGD","lr0":0.003,"momentum":0.9,"weight_decay":0.0},"mean_anchor_val_map":0.7741978751025254},{"candidate_id":"ba12245969f5","hyperparameters":{"optimizer":"MuSGD","lr0":0.01,"momentum":0.9,"weight_decay":0.0},"mean_anchor_val_map":0.7058641179474341},{"candidate_id":"34a1743a1cd9","hyperparameters":{"optimizer":"MuSGD","lr0":0.01,"momentum":0.9,"weight_decay":0.0005},"mean_anchor_val_map":0.7014966492199453},{"candidate_id":"77d398fe1ac2","hyperparameters":{"optimizer":"MuSGD","lr0":0.03,"momentum":0.9,"weight_decay":0.0005},"mean_anchor_val_map":0.3537144861118887},{"candidate_id":"62e5c47f3a75","hyperparameters":{"optimizer":"MuSGD","lr0":0.03,"momentum":0.9,"weight_decay":0.0},"mean_anchor_val_map":0.2591399398821035}],"selection_margin_ap_points":0.4639135503430314,"tried_candidates":6,"valid_candidates":6},"SGD":{"ranked":[{"candidate_id":"27eba7320c25","hyperparameters":{"optimizer":"SGD","lr0":0.003,"momentum":0.9,"weight_decay":0.0},"mean_anchor_val_map":0.7865931805383524},{"candidate_id":"4b6e48b45621","hyperparameters":{"optimizer":"SGD","lr0":0.003,"momentum":0.9,"weight_decay":0.0005},"mean_anchor_val_map":0.7774796572838193},{"candidate_id":"12d5e422d264","hyperparameters":{"optimizer":"SGD","lr0":0.01,"momentum":0.9,"weight_decay":0.0},"mean_anchor_val_map":0.7484361984924495},{"candidate_id":"7beee6af13ee","hyperparameters":{"optimizer":"SGD","lr0":0.01,"momentum":0.9,"weight_decay":0.0005},"mean_anchor_val_map":0.7315808944075719},{"candidate_id":"96be94f087f4","hyperparameters":{"optimizer":"SGD","lr0":0.03,"momentum":0.9,"weight_decay":0.0},"mean_anchor_val_map":0.49758357707873224},{"candidate_id":"71f348f7fc67","hyperparameters":{"optimizer":"SGD","lr0":0.03,"momentum":0.9,"weight_decay":0.0005},"mean_anchor_val_map":0.46732040443085987}],"selection_margin_ap_points":0.9113523254533074,"tried_candidates":6,"valid_candidates":6}},"selection_split":"wildlife validation anchors only","frozen_utc":"2026-10-25T01:39:19.857693+00:00","boundary_extension_triggered":true},"histories":[{"dataset":"wildlife","optimizer":"SGD","block":101,"fraction":0.2,"missing":0.0,"exposure":[1024,2048,3072,4096,5120,6144,7168,8192,9216,10240,11264,12288,13312,14336,15360,16384,17408,18432,19456,20480],"ap":[55.822,60.245000000000005,65.65299999999999,67.701,71.609,72.658,73.359,70.73,74.095,73.706,75.799,74.81700000000001,75.165,76.593,76.628,76.55199999999999,76.761,76.941,77.61,77.59899999999999],"box_loss":[0.77923,0.79753,0.73869,0.71477,0.71711,0.65187,0.6441,0.6199,0.59394,0.57101,0.56954,0.53822,0.51515,0.50693,0.48613,0.48088,0.44621,0.46491,0.41687,0.41865]},{"dataset":"wildlife","optimizer":"MuSGD","block":101,"fraction":0.2,"missing":0.0,"exposure":[1024,2048,3072,4096,5120,6144,7168,8192,9216,10240,11264,12288,13312,14336,15360,16384,17408,18432,19456,20480],"ap":[49.912,59.801,65.66499999999999,70.801,71.93,70.136,73.564,72.464,71.97099999999999,73.15,72.697,74.78200000000001,75.479,76.387,76.172,76.657,75.99900000000001,76.727,76.44800000000001,76.577],"box_loss":[0.83939,0.81981,0.76573,0.71701,0.70433,0.64525,0.61695,0.58867,0.56458,0.54705,0.51819,0.49149,0.47015,0.46432,0.43502,0.43033,0.39638,0.40464,0.35252,0.35434]},{"dataset":"wildlife","optimizer":"AdamW","block":101,"fraction":0.2,"missing":0.0,"exposure":[1024,2048,3072,4096,5120,6144,7168,8192,9216,10240,11264,12288,13312,14336,15360,16384,17408,18432,19456,20480],"ap":[54.867,69.694,72.14,67.656,73.416,74.287,73.682,75.356,75.676,75.902,78.12,77.534,76.946,77.13,77.535,77.012,77.347,77.077,77.473,77.493],"box_loss":[0.84656,0.79896,0.7382,0.67416,0.6797,0.62549,0.59349,0.5714,0.53703,0.50172,0.48765,0.45339,0.44153,0.43551,0.40322,0.40047,0.38005,0.3889,0.34284,0.3614]},{"dataset":"wildlife","optimizer":"AdamW","block":101,"fraction":0.2,"missing":0.2,"exposure":[1024,2048,3072,4096,5120,6144,7168,8192,9216,10240,11264,12288,13312,14336,15360,16384,17408,18432,19456,20480],"ap":[61.94499999999999,61.91,67.729,64.887,67.797,68.265,69.292,70.75200000000001,71.812,67.63900000000001,71.59,70.896,72.573,72.464,73.015,73.868,73.343,73.909,73.681,73.79299999999999],"box_loss":[0.86093,0.79363,0.7655,0.69928,0.69671,0.6276,0.61301,0.5614,0.52346,0.50286,0.48014,0.4534,0.43284,0.42664,0.41142,0.39373,0.37575,0.38048,0.33918,0.34253]},{"dataset":"wildlife","optimizer":"MuSGD","block":101,"fraction":0.2,"missing":0.2,"exposure":[1024,2048,3072,4096,5120,6144,7168,8192,9216,10240,11264,12288,13312,14336,15360,16384,17408,18432,19456,20480],"ap":[50.796,54.643,63.61599999999999,64.35600000000001,64.887,68.82300000000001,64.748,66.847,67.62599999999999,64.122,69.018,69.78,69.443,69.45400000000001,71.385,69.815,69.069,70.314,71.14399999999999,71.41300000000001],"box_loss":[0.84167,0.80642,0.76788,0.71022,0.70604,0.66019,0.63034,0.60613,0.57823,0.55445,0.52869,0.50567,0.47755,0.46438,0.44685,0.43613,0.39717,0.39654,0.34582,0.34667]},{"dataset":"wildlife","optimizer":"SGD","block":101,"fraction":0.2,"missing":0.2,"exposure":[1024,2048,3072,4096,5120,6144,7168,8192,9216,10240,11264,12288,13312,14336,15360,16384,17408,18432,19456,20480],"ap":[62.87199999999999,63.29900000000001,61.169,64.327,69.794,70.005,70.594,67.829,71.37,71.21,71.625,72.298,72.099,73.06099999999999,73.32499999999999,74.05000000000001,73.32,73.727,74.288,74.25500000000001],"box_loss":[0.80026,0.77939,0.74493,0.70931,0.69557,0.64176,0.62824,0.62062,0.59883,0.56704,0.55432,0.51645,0.51111,0.50535,0.48127,0.46961,0.43931,0.43643,0.39685,0.39347]},{"dataset":"wildlife","optimizer":"SGD","block":101,"fraction":1.0,"missing":0.0,"exposure":[1024,2048,3072,4096,5120,6144,7168,8192,9216,10240,11264,12288,13312,14336,15360,16384,17408,18432,19456,20480],"ap":[61.719,70.19999999999999,70.00999999999999,70.598,74.123,72.75399999999999,75.346,76.625,78.50200000000001,79.617,80.052,79.323,81.378,81.219,81.267,82.208,82.742,82.623,82.572,82.953],"box_loss":[0.79617,0.84108,0.82964,0.79103,0.79105,0.76331,0.76963,0.75061,0.74136,0.74157,0.70459,0.68712,0.67239,0.67957,0.63451,0.64575,0.61369,0.60237,0.5853,0.58873]},{"dataset":"wildlife","optimizer":"MuSGD","block":101,"fraction":1.0,"missing":0.0,"exposure":[1024,2048,3072,4096,5120,6144,7168,8192,9216,10240,11264,12288,13312,14336,15360,16384,17408,18432,19456,20480],"ap":[60.804,65.286,68.297,69.228,69.004,73.30499999999999,72.175,75.076,76.717,78.425,77.105,77.92999999999999,80.239,79.937,80.108,81.309,81.40899999999999,81.496,82.041,81.82199999999999],"box_loss":[0.85905,0.89824,0.8619,0.8433,0.83839,0.78825,0.78097,0.7654,0.75554,0.74517,0.69398,0.69027,0.67391,0.66979,0.62072,0.62548,0.60017,0.58949,0.56405,0.55877]},{"dataset":"wildlife","optimizer":"AdamW","block":101,"fraction":1.0,"missing":0.0,"exposure":[1024,2048,3072,4096,5120,6144,7168,8192,9216,10240,11264,12288,13312,14336,15360,16384,17408,18432,19456,20480],"ap":[58.425000000000004,67.328,73.114,74.3,72.217,74.563,73.744,77.945,77.95400000000001,79.45400000000001,79.002,80.109,80.401,80.716,81.965,81.541,82.479,82.265,82.498,82.542],"box_loss":[0.87631,0.8517,0.82744,0.80778,0.78783,0.79267,0.74828,0.73138,0.70623,0.69027,0.67316,0.64692,0.62926,0.62276,0.57791,0.59648,0.571,0.56896,0.54649,0.54487]},{"dataset":"wildlife","optimizer":"AdamW","block":101,"fraction":1.0,"missing":0.2,"exposure":[1024,2048,3072,4096,5120,6144,7168,8192,9216,10240,11264,12288,13312,14336,15360,16384,17408,18432,19456,20480],"ap":[57.598000000000006,64.827,69.37,70.015,70.697,73.462,73.475,74.66000000000001,77.42999999999999,76.629,77.347,78.47800000000001,79.483,79.704,79.186,79.281,79.574,80.032,80.408,80.29899999999999],"box_loss":[0.86482,0.84306,0.86519,0.82365,0.82735,0.77648,0.75997,0.73827,0.72928,0.70381,0.67856,0.66425,0.63529,0.62676,0.57724,0.59469,0.57252,0.57681,0.54463,0.55691]},{"dataset":"wildlife","optimizer":"MuSGD","block":101,"fraction":1.0,"missing":0.2,"exposure":[1024,2048,3072,4096,5120,6144,7168,8192,9216,10240,11264,12288,13312,14336,15360,16384,17408,18432,19456,20480],"ap":[57.208999999999996,58.043,67.407,66.956,68.49300000000001,70.765,76.535,72.946,76.494,74.71600000000001,76.767,77.078,78.521,78.033,79.01,79.861,79.679,80.686,80.538,80.734],"box_loss":[0.83736,0.89396,0.88791,0.82663,0.82233,0.82891,0.81002,0.78763,0.75899,0.75774,0.71124,0.7148,0.68109,0.67812,0.64165,0.62942,0.59701,0.59064,0.56444,0.55848]},{"dataset":"wildlife","optimizer":"SGD","block":101,"fraction":1.0,"missing":0.2,"exposure":[1024,2048,3072,4096,5120,6144,7168,8192,9216,10240,11264,12288,13312,14336,15360,16384,17408,18432,19456,20480],"ap":[56.083000000000006,63.369,68.863,70.378,67.772,73.949,74.49799999999999,76.783,77.792,76.416,78.18,78.416,79.99300000000001,79.762,80.27,80.415,81.08999999999999,81.318,81.499,81.333],"box_loss":[0.78568,0.83061,0.84664,0.78918,0.81738,0.80117,0.78142,0.75733,0.7369,0.75441,0.72474,0.70889,0.69482,0.68835,0.65378,0.64442,0.6217,0.6201,0.59946,0.57745]},{"dataset":"wildlife","optimizer":"AdamW","block":102,"fraction":0.2,"missing":0.0,"exposure":[1024,2048,3072,4096,5120,6144,7168,8192,9216,10240,11264,12288,13312,14336,15360,16384,17408,18432,19456,20480],"ap":[65.274,67.334,71.514,68.904,70.234,69.28,72.113,74.959,75.738,74.664,75.71799999999999,75.036,77.43100000000001,77.75800000000001,78.054,78.29899999999999,78.039,78.15299999999999,78.02900000000001,78.063],"box_loss":[0.85923,0.83015,0.74152,0.68602,0.66364,0.59747,0.58676,0.52699,0.52564,0.48923,0.46668,0.44343,0.42981,0.41739,0.40228,0.38455,0.36581,0.36932,0.35726,0.34567]},{"dataset":"wildlife","optimizer":"MuSGD","block":102,"fraction":0.2,"missing":0.0,"exposure":[1024,2048,3072,4096,5120,6144,7168,8192,9216,10240,11264,12288,13312,14336,15360,16384,17408,18432,19456,20480],"ap":[40.159,63.693999999999996,69.002,70.55799999999999,69.866,69.105,70.63000000000001,72.13000000000001,73.307,73.868,73.11,73.11800000000001,74.91799999999999,75.421,75.402,75.19500000000001,75.887,76.209,76.295,76.291],"box_loss":[0.82986,0.80565,0.73944,0.69915,0.69115,0.65263,0.62088,0.59046,0.58152,0.53919,0.50309,0.49202,0.486,0.458,0.45112,0.41642,0.39532,0.38828,0.36937,0.36265]},{"dataset":"wildlife","optimizer":"SGD","block":102,"fraction":0.2,"missing":0.0,"exposure":[1024,2048,3072,4096,5120,6144,7168,8192,9216,10240,11264,12288,13312,14336,15360,16384,17408,18432,19456,20480],"ap":[68.572,65.948,67.351,69.952,69.76599999999999,71.585,72.855,72.21,73.642,74.322,76.017,76.693,76.505,77.609,77.578,77.912,78.29700000000001,78.074,78.266,78.255],"box_loss":[0.77825,0.76862,0.73493,0.67873,0.67961,0.62374,0.60581,0.56971,0.5692,0.53561,0.51148,0.50286,0.48881,0.46391,0.45375,0.43977,0.42017,0.42058,0.40376,0.39241]},{"dataset":"wildlife","optimizer":"SGD","block":102,"fraction":0.2,"missing":0.2,"exposure":[1024,2048,3072,4096,5120,6144,7168,8192,9216,10240,11264,12288,13312,14336,15360,16384,17408,18432,19456,20480],"ap":[57.8,65.813,66.29599999999999,67.392,67.423,70.742,68.17800000000001,69.634,68.622,70.997,70.78999999999999,72.06700000000001,72.639,71.16900000000001,71.621,72.13300000000001,71.944,72.22,71.77900000000001,72.214],"box_loss":[0.78204,0.79602,0.7335,0.68967,0.69091,0.63661,0.62072,0.57273,0.56648,0.53089,0.52091,0.50641,0.49029,0.46904,0.45529,0.44408,0.43001,0.41467,0.39964,0.37748]},{"dataset":"wildlife","optimizer":"MuSGD","block":102,"fraction":0.2,"missing":0.2,"exposure":[1024,2048,3072,4096,5120,6144,7168,8192,9216,10240,11264,12288,13312,14336,15360,16384,17408,18432,19456,20480],"ap":[53.577,64.736,60.901,66.464,64.214,66.461,62.597,68.679,65.61200000000001,64.064,68.73,70.19999999999999,69.718,71.596,71.066,71.145,70.896,71.003,71.084,71.359],"box_loss":[0.83649,0.81703,0.73947,0.72578,0.70668,0.64455,0.63513,0.59374,0.59242,0.53489,0.51283,0.49806,0.49028,0.46147,0.44187,0.41744,0.3941,0.38724,0.36178,0.34869]},{"dataset":"wildlife","optimizer":"AdamW","block":102,"fraction":0.2,"missing":0.2,"exposure":[1024,2048,3072,4096,5120,6144,7168,8192,9216,10240,11264,12288,13312,14336,15360,16384,17408,18432,19456,20480],"ap":[59.56,62.113,66.932,65.877,62.875,66.972,66.89399999999999,67.685,67.967,69.35,69.628,71.785,69.555,69.55799999999999,70.538,72.119,71.77900000000001,71.75200000000001,71.707,71.712],"box_loss":[0.86597,0.84779,0.74939,0.69376,0.6719,0.61256,0.58996,0.54666,0.51095,0.50066,0.45333,0.4393,0.44247,0.41059,0.38904,0.38649,0.35945,0.35843,0.34686,0.33475]},{"dataset":"wildlife","optimizer":"AdamW","block":102,"fraction":1.0,"missing":0.0,"exposure":[1024,2048,3072,4096,5120,6144,7168,8192,9216,10240,11264,12288,13312,14336,15360,16384,17408,18432,19456,20480],"ap":[66.16499999999999,70.956,71.15,74.092,75.881,74.746,78.337,76.099,78.358,78.633,80.341,79.067,80.354,80.58,81.58800000000001,81.322,81.816,82.143,82.337,82.565],"box_loss":[0.87139,0.89676,0.87429,0.8209,0.81652,0.79921,0.76578,0.72976,0.71541,0.67454,0.65506,0.63712,0.63907,0.5995,0.59444,0.56974,0.56127,0.56446,0.54127,0.54943]},{"dataset":"wildlife","optimizer":"MuSGD","block":102,"fraction":1.0,"missing":0.0,"exposure":[1024,2048,3072,4096,5120,6144,7168,8192,9216,10240,11264,12288,13312,14336,15360,16384,17408,18432,19456,20480],"ap":[65.291,66.291,68.974,71.357,73.273,72.89,76.01599999999999,72.30199999999999,78.3,76.883,77.41900000000001,78.16300000000001,80.702,78.881,80.277,81.415,81.596,81.87299999999999,82.051,82.51899999999999],"box_loss":[0.8251,0.8759,0.84749,0.83084,0.80044,0.81553,0.79942,0.77201,0.73668,0.7151,0.71247,0.69762,0.69805,0.65104,0.63892,0.61452,0.60487,0.58836,0.57013,0.56377]},{"dataset":"wildlife","optimizer":"SGD","block":102,"fraction":1.0,"missing":0.0,"exposure":[1024,2048,3072,4096,5120,6144,7168,8192,9216,10240,11264,12288,13312,14336,15360,16384,17408,18432,19456,20480],"ap":[62.424,66.553,70.134,73.263,72.679,71.48,78.39500000000001,77.236,78.223,76.912,80.216,81.023,82.428,83.018,82.41000000000001,83.325,84.05,84.11999999999999,83.86200000000001,83.91300000000001],"box_loss":[0.81479,0.82231,0.83278,0.80333,0.77198,0.79422,0.76173,0.74122,0.72209,0.70901,0.69409,0.68663,0.69213,0.6587,0.63955,0.61074,0.6058,0.60161,0.5776,0.58331]},{"dataset":"wildlife","optimizer":"SGD","block":102,"fraction":1.0,"missing":0.2,"exposure":[1024,2048,3072,4096,5120,6144,7168,8192,9216,10240,11264,12288,13312,14336,15360,16384,17408,18432,19456,20480],"ap":[63.55,65.625,64.425,69.28,73.917,68.474,76.235,75.01400000000001,77.86699999999999,76.55499999999999,77.86699999999999,78.171,78.63,78.767,79.327,80.26299999999999,80.72,80.401,80.31,80.648],"box_loss":[0.81715,0.87399,0.85319,0.81714,0.81715,0.80404,0.75892,0.74566,0.71646,0.71322,0.69023,0.68671,0.6781,0.66562,0.64944,0.62427,0.60382,0.59229,0.57352,0.56985]},{"dataset":"wildlife","optimizer":"MuSGD","block":102,"fraction":1.0,"missing":0.2,"exposure":[1024,2048,3072,4096,5120,6144,7168,8192,9216,10240,11264,12288,13312,14336,15360,16384,17408,18432,19456,20480],"ap":[61.126000000000005,66.51,61.045,66.965,69.682,69.14200000000001,74.105,75.149,73.032,72.56899999999999,75.913,76.98899999999999,76.089,77.188,78.41900000000001,78.79899999999999,79.624,79.429,79.39,79.296],"box_loss":[0.88191,0.89741,0.88298,0.86775,0.83536,0.82428,0.80623,0.8026,0.76275,0.74823,0.71514,0.71566,0.69762,0.66817,0.65308,0.61909,0.59627,0.58917,0.565,0.56532]},{"dataset":"wildlife","optimizer":"AdamW","block":102,"fraction":1.0,"missing":0.2,"exposure":[1024,2048,3072,4096,5120,6144,7168,8192,9216,10240,11264,12288,13312,14336,15360,16384,17408,18432,19456,20480],"ap":[67.952,66.251,69.344,70.855,71.611,73.355,72.725,76.639,76.594,75.672,75.614,75.94800000000001,78.31099999999999,77.032,77.602,78.29700000000001,78.652,78.745,78.89500000000001,78.745],"box_loss":[0.90118,0.9251,0.89068,0.86374,0.80449,0.79597,0.75631,0.74202,0.69529,0.66651,0.64713,0.64948,0.6338,0.61147,0.59763,0.57726,0.56423,0.55161,0.54585,0.54787]},{"dataset":"wildlife","optimizer":"SGD","block":103,"fraction":0.2,"missing":0.0,"exposure":[1024,2048,3072,4096,5120,6144,7168,8192,9216,10240,11264,12288,13312,14336,15360,16384,17408,18432,19456,20480],"ap":[60.112,58.744,66.086,69.07600000000001,70.906,69.396,68.269,71.83500000000001,73.151,74.33,74.722,76.301,77.125,77.271,76.91900000000001,77.697,77.883,77.339,77.545,77.009],"box_loss":[0.72586,0.70891,0.70673,0.67309,0.6371,0.63835,0.58742,0.5759,0.54921,0.53845,0.50911,0.48569,0.46656,0.44133,0.43328,0.40713,0.40924,0.40576,0.38366,0.37034]},{"dataset":"wildlife","optimizer":"MuSGD","block":103,"fraction":0.2,"missing":0.0,"exposure":[1024,2048,3072,4096,5120,6144,7168,8192,9216,10240,11264,12288,13312,14336,15360,16384,17408,18432,19456,20480],"ap":[49.395,59.70100000000001,62.097,63.260000000000005,69.28800000000001,68.11500000000001,68.281,72.409,72.407,74.815,72.22,74.22800000000001,73.994,75.351,75.703,75.775,75.912,76.545,76.52799999999999,76.21],"box_loss":[0.79206,0.76046,0.7047,0.68682,0.65539,0.62484,0.60113,0.5706,0.53594,0.5263,0.50966,0.46601,0.44728,0.41898,0.40408,0.38179,0.38223,0.36607,0.34936,0.32445]},{"dataset":"wildlife","optimizer":"AdamW","block":103,"fraction":0.2,"missing":0.0,"exposure":[1024,2048,3072,4096,5120,6144,7168,8192,9216,10240,11264,12288,13312,14336,15360,16384,17408,18432,19456,20480],"ap":[57.113,64.184,70.028,69.801,68.387,71.133,71.405,72.541,74.259,73.463,73.262,75.12,74.248,74.65,75.24,75.522,75.18599999999999,75.453,75.517,75.223],"box_loss":[0.83629,0.76196,0.7126,0.67231,0.62474,0.60562,0.54646,0.51231,0.49652,0.47624,0.4518,0.41735,0.4081,0.38782,0.36858,0.35228,0.34906,0.35195,0.33453,0.32253]},{"dataset":"wildlife","optimizer":"AdamW","block":103,"fraction":0.2,"missing":0.2,"exposure":[1024,2048,3072,4096,5120,6144,7168,8192,9216,10240,11264,12288,13312,14336,15360,16384,17408,18432,19456,20480],"ap":[56.062999999999995,65.593,67.728,68.25099999999999,64.863,67.499,67.74,69.973,70.25200000000001,71.467,71.938,71.281,71.26400000000001,71.458,71.851,72.409,72.575,72.773,72.881,72.993],"box_loss":[0.8143,0.78082,0.72175,0.65169,0.6012,0.59918,0.54746,0.5238,0.48154,0.46448,0.44438,0.41982,0.40522,0.36906,0.35642,0.34119,0.33145,0.34169,0.32324,0.30223]},{"dataset":"wildlife","optimizer":"MuSGD","block":103,"fraction":0.2,"missing":0.2,"exposure":[1024,2048,3072,4096,5120,6144,7168,8192,9216,10240,11264,12288,13312,14336,15360,16384,17408,18432,19456,20480],"ap":[55.30199999999999,58.132,60.007999999999996,59.559,66.44099999999999,64.864,68.999,66.719,69.742,68.16799999999999,68.05600000000001,69.962,69.83,69.666,70.592,71.709,69.821,71.121,70.89999999999999,71.147],"box_loss":[0.74945,0.77547,0.74147,0.68673,0.65285,0.61117,0.58102,0.57932,0.533,0.53019,0.50778,0.47707,0.44981,0.41986,0.39825,0.38643,0.36779,0.3635,0.34712,0.31869]},{"dataset":"wildlife","optimizer":"SGD","block":103,"fraction":0.2,"missing":0.2,"exposure":[1024,2048,3072,4096,5120,6144,7168,8192,9216,10240,11264,12288,13312,14336,15360,16384,17408,18432,19456,20480],"ap":[56.233,58.645,66.63900000000001,66.979,66.685,67.566,68.833,67.781,70.062,72.194,71.076,69.62,71.332,70.638,71.029,71.19,71.344,71.783,72.31,71.885],"box_loss":[0.73795,0.72498,0.68727,0.65127,0.63537,0.60862,0.57045,0.56619,0.52129,0.51283,0.50607,0.47196,0.45821,0.43225,0.42962,0.3966,0.39724,0.40717,0.38009,0.35175]},{"dataset":"wildlife","optimizer":"SGD","block":103,"fraction":1.0,"missing":0.0,"exposure":[1024,2048,3072,4096,5120,6144,7168,8192,9216,10240,11264,12288,13312,14336,15360,16384,17408,18432,19456,20480],"ap":[62.197,69.367,71.72800000000001,71.696,70.788,73.929,74.53,74.809,77.558,77.916,78.361,78.293,79.47,80.132,80.511,80.669,81.144,81.173,81.355,81.605],"box_loss":[0.77693,0.82866,0.82051,0.80623,0.78651,0.77688,0.7624,0.75705,0.74154,0.71254,0.68481,0.69136,0.64685,0.64183,0.62413,0.62856,0.59188,0.60352,0.58658,0.57645]},{"dataset":"wildlife","optimizer":"MuSGD","block":103,"fraction":1.0,"missing":0.0,"exposure":[1024,2048,3072,4096,5120,6144,7168,8192,9216,10240,11264,12288,13312,14336,15360,16384,17408,18432,19456,20480],"ap":[53.254000000000005,54.690000000000005,67.243,68.118,73.873,73.925,75.057,73.71600000000001,73.616,76.377,77.53,76.598,80.205,78.62700000000001,80.326,81.326,81.404,81.854,82.12,81.99600000000001],"box_loss":[0.82797,0.86212,0.85421,0.82665,0.82972,0.79552,0.79203,0.75038,0.73275,0.73481,0.69625,0.69607,0.65017,0.65926,0.62782,0.62533,0.58343,0.58656,0.57316,0.54886]},{"dataset":"wildlife","optimizer":"AdamW","block":103,"fraction":1.0,"missing":0.0,"exposure":[1024,2048,3072,4096,5120,6144,7168,8192,9216,10240,11264,12288,13312,14336,15360,16384,17408,18432,19456,20480],"ap":[56.792,66.918,74.179,73.02199999999999,75.103,75.345,76.415,75.291,78.021,77.605,78.943,78.83,79.686,81.829,80.77,81.358,80.808,81.06700000000001,81.108,81.32300000000001],"box_loss":[0.87392,0.88606,0.84823,0.82778,0.78357,0.77782,0.74613,0.72477,0.70276,0.69553,0.65853,0.66066,0.60982,0.60654,0.57863,0.58851,0.55867,0.57399,0.56152,0.54508]},{"dataset":"wildlife","optimizer":"AdamW","block":103,"fraction":1.0,"missing":0.2,"exposure":[1024,2048,3072,4096,5120,6144,7168,8192,9216,10240,11264,12288,13312,14336,15360,16384,17408,18432,19456,20480],"ap":[58.619,66.24799999999999,70.844,68.691,72.653,72.626,73.425,74.115,75.57199999999999,73.272,77.203,77.648,77.74600000000001,78.766,78.732,78.28,78.729,78.86,79.324,79.163],"box_loss":[0.90638,0.91195,0.87306,0.82013,0.82095,0.78441,0.75887,0.7301,0.71247,0.68062,0.64547,0.6464,0.60729,0.60629,0.58803,0.59256,0.55564,0.56853,0.55873,0.54884]},{"dataset":"wildlife","optimizer":"MuSGD","block":103,"fraction":1.0,"missing":0.2,"exposure":[1024,2048,3072,4096,5120,6144,7168,8192,9216,10240,11264,12288,13312,14336,15360,16384,17408,18432,19456,20480],"ap":[62.846000000000004,61.266,64.986,70.102,70.048,71.17,71.05900000000001,69.56400000000001,74.46000000000001,72.32,76.25999999999999,75.103,77.366,79.231,78.51700000000001,78.168,79.032,79.769,79.78,79.542],"box_loss":[0.84891,0.89668,0.89434,0.87342,0.85427,0.82736,0.80022,0.79376,0.78356,0.73726,0.72171,0.71036,0.6798,0.68041,0.63628,0.63843,0.59609,0.59712,0.58565,0.54751]},{"dataset":"wildlife","optimizer":"SGD","block":103,"fraction":1.0,"missing":0.2,"exposure":[1024,2048,3072,4096,5120,6144,7168,8192,9216,10240,11264,12288,13312,14336,15360,16384,17408,18432,19456,20480],"ap":[53.82900000000001,67.32900000000001,65.28,68.399,73.615,73.82,73.97,73.559,74.208,74.976,77.781,76.5,77.606,77.89,78.718,78.53,79.099,79.337,79.484,79.682],"box_loss":[0.82116,0.8345,0.85937,0.80917,0.81245,0.77767,0.77579,0.76078,0.73135,0.71945,0.69929,0.7056,0.65475,0.65293,0.6282,0.62856,0.59206,0.60971,0.59235,0.56892]},{"dataset":"voc2007","optimizer":"MuSGD","block":101,"fraction":0.2,"missing":0.0,"exposure":[1024,2048,3072,4096,5120,6144,7168,8192,9216,10240,11264,12288,13312,14336,15360,16384,17408,18432,19456,20480],"ap":[32.704,37.103,38.376,39.483000000000004,40.939,40.841,42.439,45.202,43.988,45.86,46.269,46.53,47.173,47.758,47.53,48.569,49.263,49.534,49.894,49.915],"box_loss":[1.2241,1.24534,1.21901,1.1972,1.18107,1.1205,1.14153,1.11945,1.07586,1.04386,1.02158,0.97816,0.99701,0.94194,0.93481,0.91621,0.88745,0.85794,0.84416,0.83272]},{"dataset":"voc2007","optimizer":"AdamW","block":101,"fraction":0.2,"missing":0.0,"exposure":[1024,2048,3072,4096,5120,6144,7168,8192,9216,10240,11264,12288,13312,14336,15360,16384,17408,18432,19456,20480],"ap":[29.043999999999997,35.384,38.792,40.164,40.42,40.816,41.616,44.534,44.151,45.484,45.949,45.437,46.631,46.442,46.666000000000004,46.885,47.314,47.736000000000004,47.467999999999996,47.562],"box_loss":[1.29106,1.26751,1.25736,1.21464,1.17033,1.10902,1.08575,1.09026,1.03849,1.01966,0.99419,0.92008,0.95223,0.91121,0.89351,0.8806,0.85997,0.84114,0.84596,0.8261]},{"dataset":"voc2007","optimizer":"AdamW","block":101,"fraction":0.2,"missing":0.2,"exposure":[1024,2048,3072,4096,5120,6144,7168,8192,9216,10240,11264,12288,13312,14336,15360,16384,17408,18432,19456,20480],"ap":[26.593,32.909,37.325,34.861,38.086,38.98,38.551,39.795,41.385,40.536,40.802,41.287,42.132,41.943999999999996,42.487,42.657000000000004,43.141,42.981,42.989,43.125],"box_loss":[1.3001,1.2759,1.25765,1.23507,1.18988,1.12897,1.10496,1.10522,1.05567,1.02174,0.98266,0.92388,0.95827,0.88988,0.88498,0.87347,0.8611,0.84957,0.83413,0.828]},{"dataset":"voc2007","optimizer":"MuSGD","block":101,"fraction":0.2,"missing":0.2,"exposure":[1024,2048,3072,4096,5120,6144,7168,8192,9216,10240,11264,12288,13312,14336,15360,16384,17408,18432,19456,20480],"ap":[27.253,36.315999999999995,33.876,38.46,37.155,39.939,38.231,38.407000000000004,42.571,42.848000000000006,41.426,42.604,42.82,42.658,44.699,44.330999999999996,44.75,45.111000000000004,45.418,45.379000000000005],"box_loss":[1.21695,1.25603,1.25812,1.21916,1.21403,1.15103,1.15045,1.13445,1.09296,1.06316,1.02236,0.96131,1.00153,0.94057,0.93934,0.90133,0.87697,0.87081,0.83571,0.81978]},{"dataset":"voc2007","optimizer":"MuSGD","block":101,"fraction":1.0,"missing":0.0,"exposure":[1024,2048,3072,4096,5120,6144,7168,8192,9216,10240,11264,12288,13312,14336,15360,16384,17408,18432,19456,20480],"ap":[36.24,36.635,35.467,38.382,40.365,43.021,42.176,44.796,44.946999999999996,48.291000000000004,48.909000000000006,49.989,51.158,52.125,53.815000000000005,54.37800000000001,54.337,55.224,55.379,55.602],"box_loss":[1.19723,1.33015,1.28619,1.31469,1.33337,1.28752,1.27759,1.27262,1.25178,1.20647,1.22026,1.22531,1.15354,1.15677,1.19338,1.12476,1.13504,1.0841,1.04327,1.08079]},{"dataset":"voc2007","optimizer":"AdamW","block":101,"fraction":1.0,"missing":0.0,"exposure":[1024,2048,3072,4096,5120,6144,7168,8192,9216,10240,11264,12288,13312,14336,15360,16384,17408,18432,19456,20480],"ap":[28.143,36.769,39.704,40.266999999999996,41.711999999999996,45.066,45.869,47.129,47.539,50.05500000000001,50.212,51.591,52.431000000000004,52.898999999999994,53.013,54.947,54.551,54.998000000000005,55.516,55.47],"box_loss":[1.27756,1.35562,1.30709,1.31064,1.27302,1.27845,1.24933,1.24356,1.2146,1.18115,1.19079,1.17338,1.14173,1.10222,1.15035,1.08157,1.09822,1.04194,1.03383,1.057]},{"dataset":"voc2007","optimizer":"AdamW","block":101,"fraction":1.0,"missing":0.2,"exposure":[1024,2048,3072,4096,5120,6144,7168,8192,9216,10240,11264,12288,13312,14336,15360,16384,17408,18432,19456,20480],"ap":[21.598,33.611999999999995,38.613,36.782,40.177,43.968,42.535000000000004,45.834,45.486,47.0,49.091,47.955999999999996,48.907000000000004,50.391,50.516000000000005,51.68000000000001,52.295,52.727999999999994,53.064,53.08500000000001],"box_loss":[1.28077,1.35637,1.31155,1.32504,1.32043,1.27548,1.2591,1.23974,1.23409,1.20002,1.19085,1.2035,1.12984,1.12777,1.15815,1.08661,1.08803,1.064,1.02356,1.08995]},{"dataset":"voc2007","optimizer":"MuSGD","block":101,"fraction":1.0,"missing":0.2,"exposure":[1024,2048,3072,4096,5120,6144,7168,8192,9216,10240,11264,12288,13312,14336,15360,16384,17408,18432,19456,20480],"ap":[27.182000000000002,29.744,36.851,35.53,37.552,38.249,42.245,41.625,41.559000000000005,44.346000000000004,45.499,49.157000000000004,48.765,51.304,50.724999999999994,51.141999999999996,53.417,53.33599999999999,53.92400000000001,54.696],"box_loss":[1.19411,1.3545,1.31892,1.34779,1.34354,1.30585,1.29391,1.29025,1.28532,1.22508,1.24936,1.22724,1.1808,1.18592,1.21433,1.14783,1.14194,1.10242,1.05447,1.10251]},{"dataset":"voc2007","optimizer":"AdamW","block":102,"fraction":0.2,"missing":0.0,"exposure":[1024,2048,3072,4096,5120,6144,7168,8192,9216,10240,11264,12288,13312,14336,15360,16384,17408,18432,19456,20480],"ap":[30.36,35.65,40.193,38.265,41.998000000000005,42.199999999999996,43.183,43.010999999999996,42.529,43.428,44.668,44.79,44.958,46.007999999999996,47.018,46.499,47.624,47.491,47.650999999999996,47.707],"box_loss":[1.20923,1.21543,1.20856,1.15817,1.1296,1.12146,1.08311,1.0639,1.05807,0.9986,0.97905,0.94329,0.95506,0.91178,0.89337,0.88097,0.85167,0.83037,0.84208,0.8178]},{"dataset":"voc2007","optimizer":"MuSGD","block":102,"fraction":0.2,"missing":0.0,"exposure":[1024,2048,3072,4096,5120,6144,7168,8192,9216,10240,11264,12288,13312,14336,15360,16384,17408,18432,19456,20480],"ap":[31.411,38.301,39.828,40.602,41.608000000000004,43.657000000000004,39.972,42.667,43.299,44.394,45.388,43.484,45.275999999999996,47.199000000000005,48.373,48.364000000000004,49.812,49.559,49.741,49.611],"box_loss":[1.18531,1.22149,1.21073,1.17168,1.14109,1.14408,1.10923,1.082,1.06905,1.02229,1.01495,0.98061,0.98364,0.94911,0.92139,0.89502,0.86239,0.83387,0.84909,0.81612]},{"dataset":"voc2007","optimizer":"MuSGD","block":102,"fraction":0.2,"missing":0.2,"exposure":[1024,2048,3072,4096,5120,6144,7168,8192,9216,10240,11264,12288,13312,14336,15360,16384,17408,18432,19456,20480],"ap":[33.444,36.301,33.754,32.413,37.495,35.199999999999996,40.147,40.711000000000006,41.187000000000005,41.144,42.397,43.619,43.156,43.681,44.974,44.602000000000004,44.767,45.437,45.539,45.617999999999995],"box_loss":[1.20203,1.26296,1.23508,1.19713,1.15175,1.16895,1.09643,1.09303,1.07505,1.03673,1.02241,0.96237,0.9809,0.93996,0.9224,0.9054,0.86239,0.8273,0.8438,0.81108]},{"dataset":"voc2007","optimizer":"AdamW","block":102,"fraction":0.2,"missing":0.2,"exposure":[1024,2048,3072,4096,5120,6144,7168,8192,9216,10240,11264,12288,13312,14336,15360,16384,17408,18432,19456,20480],"ap":[29.352,34.336,38.128,36.697,38.41,38.234,39.214,39.81,38.85,41.064,40.47,41.281,41.705,42.931999999999995,42.546,42.473,43.105,43.45,43.589,43.297999999999995],"box_loss":[1.25417,1.24453,1.24001,1.19152,1.15098,1.13203,1.10027,1.05508,1.0572,1.01256,0.98842,0.94453,0.95086,0.92644,0.89161,0.87697,0.85278,0.82673,0.84225,0.81995]},{"dataset":"voc2007","optimizer":"AdamW","block":102,"fraction":1.0,"missing":0.0,"exposure":[1024,2048,3072,4096,5120,6144,7168,8192,9216,10240,11264,12288,13312,14336,15360,16384,17408,18432,19456,20480],"ap":[29.494,35.891,37.95,40.991,43.245,44.955,45.132,45.028,48.246,49.195,48.338,50.43,51.749,53.476,54.730999999999995,55.227000000000004,54.781,55.106,55.086,55.262],"box_loss":[1.2321,1.30166,1.30514,1.29755,1.26096,1.23427,1.2348,1.25311,1.23044,1.1767,1.1186,1.16276,1.14997,1.11954,1.09479,1.09263,1.06511,1.04319,1.06525,1.08923]},{"dataset":"voc2007","optimizer":"MuSGD","block":102,"fraction":1.0,"missing":0.0,"exposure":[1024,2048,3072,4096,5120,6144,7168,8192,9216,10240,11264,12288,13312,14336,15360,16384,17408,18432,19456,20480],"ap":[34.379,35.497,39.803,40.605000000000004,41.789,42.181000000000004,44.529999999999994,44.265,48.345,47.658,49.265,48.892,52.322,52.672,53.029,54.681000000000004,54.272,55.159000000000006,55.576,55.667],"box_loss":[1.16438,1.32392,1.30907,1.30592,1.28943,1.25003,1.26294,1.29675,1.23808,1.22751,1.1752,1.20877,1.18475,1.14934,1.15901,1.12752,1.11447,1.07828,1.09572,1.0984]},{"dataset":"voc2007","optimizer":"MuSGD","block":102,"fraction":1.0,"missing":0.2,"exposure":[1024,2048,3072,4096,5120,6144,7168,8192,9216,10240,11264,12288,13312,14336,15360,16384,17408,18432,19456,20480],"ap":[26.112999999999996,34.726,37.272,39.249,41.52,41.728,44.374,43.3,45.354,45.793,48.55,48.433,51.407000000000004,51.596,52.83,52.599,53.529,54.259,54.638,55.142],"box_loss":[1.17722,1.31973,1.33958,1.30272,1.30188,1.28397,1.30084,1.2896,1.29077,1.24852,1.21862,1.21195,1.21702,1.17899,1.15237,1.14176,1.1087,1.08954,1.10677,1.10072]},{"dataset":"voc2007","optimizer":"AdamW","block":102,"fraction":1.0,"missing":0.2,"exposure":[1024,2048,3072,4096,5120,6144,7168,8192,9216,10240,11264,12288,13312,14336,15360,16384,17408,18432,19456,20480],"ap":[26.203,33.804,37.413999999999994,41.522,41.556,44.025,45.122,44.905,46.914,47.325,48.239,49.038,49.968,50.849,50.761,51.76,51.791,52.414,52.55800000000001,52.608999999999995],"box_loss":[1.25259,1.2919,1.31442,1.30154,1.29059,1.26845,1.2517,1.27151,1.24576,1.21841,1.16693,1.17471,1.17668,1.12737,1.11516,1.10157,1.07002,1.06713,1.07888,1.08312]},{"dataset":"voc2007","optimizer":"MuSGD","block":103,"fraction":0.2,"missing":0.0,"exposure":[1024,2048,3072,4096,5120,6144,7168,8192,9216,10240,11264,12288,13312,14336,15360,16384,17408,18432,19456,20480],"ap":[36.148,36.44,39.491,40.306999999999995,40.524,41.258,39.729,40.744,43.072,43.282,44.767,45.828,44.975,46.397,47.647,47.407,47.399,47.58,48.39,48.419000000000004],"box_loss":[1.17482,1.19207,1.20529,1.16398,1.1475,1.12888,1.08522,1.09229,1.05885,1.0189,1.0246,0.97406,0.9458,0.95005,0.8966,0.8954,0.88121,0.86564,0.81793,0.82201]},{"dataset":"voc2007","optimizer":"AdamW","block":103,"fraction":0.2,"missing":0.0,"exposure":[1024,2048,3072,4096,5120,6144,7168,8192,9216,10240,11264,12288,13312,14336,15360,16384,17408,18432,19456,20480],"ap":[29.815,35.554,39.535,41.298,40.818,42.766999999999996,43.105,41.034,43.32,43.265,44.107,44.876,45.234,45.545,46.686,46.853,46.589999999999996,47.131,47.321999999999996,47.587],"box_loss":[1.2425,1.23705,1.20512,1.16902,1.14219,1.09709,1.08744,1.06151,1.03091,0.99279,0.99792,0.9457,0.90046,0.89501,0.87404,0.86635,0.86301,0.85484,0.82077,0.83249]},{"dataset":"voc2007","optimizer":"AdamW","block":103,"fraction":0.2,"missing":0.2,"exposure":[1024,2048,3072,4096,5120,6144,7168,8192,9216,10240,11264,12288,13312,14336,15360,16384,17408,18432,19456,20480],"ap":[26.929,33.461999999999996,37.403,39.175,39.153,38.007000000000005,40.811,41.209,40.21,40.515,41.67,41.894999999999996,41.775,42.437999999999995,42.388,43.017,43.309999999999995,43.312,43.605,43.635000000000005],"box_loss":[1.21854,1.23354,1.22349,1.16177,1.12574,1.11566,1.0893,1.04747,1.02924,0.98688,0.98228,0.93156,0.89386,0.88633,0.85978,0.85597,0.84814,0.84418,0.79941,0.80274]},{"dataset":"voc2007","optimizer":"MuSGD","block":103,"fraction":0.2,"missing":0.2,"exposure":[1024,2048,3072,4096,5120,6144,7168,8192,9216,10240,11264,12288,13312,14336,15360,16384,17408,18432,19456,20480],"ap":[30.37,33.969,35.709999999999994,33.671,37.7,37.2,36.968,38.466,39.355000000000004,40.997,41.415,41.678,42.781000000000006,44.04,44.003,44.829,44.407000000000004,44.817,44.864,44.988],"box_loss":[1.15603,1.21793,1.2145,1.17847,1.15057,1.12996,1.11425,1.04642,1.03866,1.01499,1.02282,0.97043,0.93811,0.93838,0.89614,0.87378,0.86455,0.85589,0.80445,0.79698]},{"dataset":"voc2007","optimizer":"MuSGD","block":103,"fraction":1.0,"missing":0.0,"exposure":[1024,2048,3072,4096,5120,6144,7168,8192,9216,10240,11264,12288,13312,14336,15360,16384,17408,18432,19456,20480],"ap":[37.994,38.233,38.798,36.788,41.374,41.099999999999994,42.067,45.574,43.042,47.620000000000005,45.296,50.381,51.217,52.503,51.739000000000004,53.900999999999996,53.925999999999995,54.30500000000001,55.24,55.11899999999999],"box_loss":[1.14751,1.29049,1.31472,1.30337,1.29875,1.2716,1.24242,1.25177,1.20071,1.22351,1.18762,1.22347,1.2081,1.14824,1.1152,1.14223,1.11268,1.08735,1.06356,1.08267]},{"dataset":"voc2007","optimizer":"AdamW","block":103,"fraction":1.0,"missing":0.0,"exposure":[1024,2048,3072,4096,5120,6144,7168,8192,9216,10240,11264,12288,13312,14336,15360,16384,17408,18432,19456,20480],"ap":[31.480999999999998,34.897,40.507,40.684,43.216,43.053999999999995,46.654,47.314,47.739,48.896,49.943,51.048,51.437,52.317,52.335,53.476,53.596999999999994,54.044000000000004,53.727999999999994,54.22],"box_loss":[1.22192,1.29872,1.31106,1.29926,1.29177,1.26274,1.24043,1.23719,1.17536,1.21117,1.15466,1.17399,1.17958,1.10289,1.09892,1.10794,1.07274,1.05819,1.05582,1.07656]},{"dataset":"voc2007","optimizer":"AdamW","block":103,"fraction":1.0,"missing":0.2,"exposure":[1024,2048,3072,4096,5120,6144,7168,8192,9216,10240,11264,12288,13312,14336,15360,16384,17408,18432,19456,20480],"ap":[27.683000000000003,34.801,38.41,39.528999999999996,41.974000000000004,44.169000000000004,42.457,45.184999999999995,45.792,47.760000000000005,49.553000000000004,48.882,50.111000000000004,50.676,51.562,52.198,52.659,52.858000000000004,53.016,53.288000000000004],"box_loss":[1.2328,1.34141,1.36036,1.29908,1.30279,1.27459,1.24401,1.24117,1.19597,1.23172,1.17385,1.18692,1.19271,1.11426,1.11707,1.11881,1.08312,1.07525,1.08181,1.08891]},{"dataset":"voc2007","optimizer":"MuSGD","block":103,"fraction":1.0,"missing":0.2,"exposure":[1024,2048,3072,4096,5120,6144,7168,8192,9216,10240,11264,12288,13312,14336,15360,16384,17408,18432,19456,20480],"ap":[32.800000000000004,36.956,36.353,38.667,39.596,42.903999999999996,39.581,41.724,45.256,47.143,43.185,46.592,49.036,51.056999999999995,51.07000000000001,51.639,52.197,53.493,53.647,53.869],"box_loss":[1.15925,1.32688,1.33455,1.33231,1.30839,1.27771,1.28442,1.28203,1.25094,1.25071,1.21153,1.23061,1.24937,1.18527,1.14305,1.15293,1.10984,1.10018,1.09233,1.09334]}],"trials":[{"run_id":"journal-wildlife-AdamW-s42-43716c611b6a","dataset":"wildlife","optimizer":"AdamW","seed":42,"data_fraction":0.2,"noise_probability":0.0,"realized_noise_rate":0.0,"lr0":0.001,"train_images":204,"images_seen":4160,"optimizer_steps":260,"epochs_completed":20,"training_seconds":118.94472989500002,"val_map50_95":0.699993833170178,"roles":"variable_exposure_control"},{"run_id":"journal-wildlife-MuSGD-s42-c1be1dfa00f4","dataset":"wildlife","optimizer":"MuSGD","seed":42,"data_fraction":0.2,"noise_probability":0.0,"realized_noise_rate":0.0,"lr0":0.01,"train_images":204,"images_seen":4160,"optimizer_steps":260,"epochs_completed":20,"training_seconds":92.09538518700003,"val_map50_95":0.6984273616162175,"roles":"variable_exposure_control"},{"run_id":"journal-wildlife-MuSGD-s42-80cc28a85686","dataset":"wildlife","optimizer":"MuSGD","seed":42,"data_fraction":0.2,"noise_probability":0.2,"realized_noise_rate":0.1875,"lr0":0.01,"train_images":204,"images_seen":4160,"optimizer_steps":260,"epochs_completed":20,"training_seconds":94.15632479299995,"val_map50_95":0.6633104270134675,"roles":"variable_exposure_control"},{"run_id":"journal-wildlife-AdamW-s42-8a5f4286e981","dataset":"wildlife","optimizer":"AdamW","seed":42,"data_fraction":0.2,"noise_probability":0.2,"realized_noise_rate":0.1875,"lr0":0.001,"train_images":204,"images_seen":4160,"optimizer_steps":260,"epochs_completed":20,"training_seconds":92.34406011099998,"val_map50_95":0.6849229925623093,"roles":"variable_exposure_control"},{"run_id":"journal-wildlife-AdamW-s42-26d6aee9e7a9","dataset":"wildlife","optimizer":"AdamW","seed":42,"data_fraction":1.0,"noise_probability":0.0,"realized_noise_rate":0.0,"lr0":0.001,"train_images":1020,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":313.92058952699995,"val_map50_95":0.7794272898926609,"roles":"variable_exposure_control | fixed_exposure_defaults"},{"run_id":"journal-wildlife-MuSGD-s42-6b446ab51f73","dataset":"wildlife","optimizer":"MuSGD","seed":42,"data_fraction":1.0,"noise_probability":0.0,"realized_noise_rate":0.0,"lr0":0.01,"train_images":1020,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":336.6929932570001,"val_map50_95":0.7666388470925829,"roles":"variable_exposure_control | fixed_exposure_defaults"},{"run_id":"journal-wildlife-MuSGD-s42-2dbd732f206a","dataset":"wildlife","optimizer":"MuSGD","seed":42,"data_fraction":1.0,"noise_probability":0.2,"realized_noise_rate":0.1920212765957447,"lr0":0.01,"train_images":1020,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":337.27295985599994,"val_map50_95":0.7527470818785102,"roles":"variable_exposure_control | fixed_exposure_defaults"},{"run_id":"journal-wildlife-AdamW-s42-f8e9a61157b1","dataset":"wildlife","optimizer":"AdamW","seed":42,"data_fraction":1.0,"noise_probability":0.2,"realized_noise_rate":0.1920212765957447,"lr0":0.001,"train_images":1020,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":315.25910837100014,"val_map50_95":0.7648202627892415,"roles":"variable_exposure_control | fixed_exposure_defaults"},{"run_id":"journal-wildlife-AdamW-s42-787130ee3c6e","dataset":"wildlife","optimizer":"AdamW","seed":42,"data_fraction":0.2,"noise_probability":0.0,"realized_noise_rate":0.0,"lr0":0.001,"train_images":204,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":313.641670084,"val_map50_95":0.7303441685494644,"roles":"fixed_exposure_defaults"},{"run_id":"journal-wildlife-MuSGD-s42-daab69f2aafc","dataset":"wildlife","optimizer":"MuSGD","seed":42,"data_fraction":0.2,"noise_probability":0.0,"realized_noise_rate":0.0,"lr0":0.01,"train_images":204,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":335.54486644200006,"val_map50_95":0.7019480565328684,"roles":"fixed_exposure_defaults"},{"run_id":"journal-wildlife-SGD-s42-289ca089d7ea","dataset":"wildlife","optimizer":"SGD","seed":42,"data_fraction":0.2,"noise_probability":0.0,"realized_noise_rate":0.0,"lr0":0.01,"train_images":204,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":311.6043963950001,"val_map50_95":0.7395286233013147,"roles":"fixed_exposure_defaults"},{"run_id":"journal-wildlife-SGD-s42-925ffdbff252","dataset":"wildlife","optimizer":"SGD","seed":42,"data_fraction":0.2,"noise_probability":0.2,"realized_noise_rate":0.1875,"lr0":0.01,"train_images":204,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":312.01788673600004,"val_map50_95":0.6789250365324441,"roles":"fixed_exposure_defaults"},{"run_id":"journal-wildlife-MuSGD-s42-495d62c794d2","dataset":"wildlife","optimizer":"MuSGD","seed":42,"data_fraction":0.2,"noise_probability":0.2,"realized_noise_rate":0.1875,"lr0":0.01,"train_images":204,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":335.2741458669998,"val_map50_95":0.6483732036783317,"roles":"fixed_exposure_defaults"},{"run_id":"journal-wildlife-AdamW-s42-3cca49362752","dataset":"wildlife","optimizer":"AdamW","seed":42,"data_fraction":0.2,"noise_probability":0.2,"realized_noise_rate":0.1875,"lr0":0.001,"train_images":204,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":314.0284820439997,"val_map50_95":0.6796095057323619,"roles":"fixed_exposure_defaults"},{"run_id":"journal-wildlife-SGD-s42-db3e43265fdb","dataset":"wildlife","optimizer":"SGD","seed":42,"data_fraction":1.0,"noise_probability":0.0,"realized_noise_rate":0.0,"lr0":0.01,"train_images":1020,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":311.6054353319996,"val_map50_95":0.798144238344396,"roles":"fixed_exposure_defaults"},{"run_id":"journal-wildlife-SGD-s42-69dae5dab330","dataset":"wildlife","optimizer":"SGD","seed":42,"data_fraction":1.0,"noise_probability":0.2,"realized_noise_rate":0.1920212765957447,"lr0":0.01,"train_images":1020,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":311.69057095599965,"val_map50_95":0.7661014809219729,"roles":"fixed_exposure_defaults"},{"run_id":"journal-wildlife-AdamW-s11-684783e670f8","dataset":"wildlife","optimizer":"AdamW","seed":11,"data_fraction":1.0,"noise_probability":0.0,"realized_noise_rate":0.0,"lr0":0.0003,"train_images":1020,"images_seen":10240,"optimizer_steps":640,"epochs_completed":10,"training_seconds":158.80858212099974,"val_map50_95":0.808606119971277,"roles":"tuning"},{"run_id":"journal-wildlife-AdamW-s11-6954e3e21e56","dataset":"wildlife","optimizer":"AdamW","seed":11,"data_fraction":0.2,"noise_probability":0.2,"realized_noise_rate":0.1875,"lr0":0.0003,"train_images":204,"images_seen":10240,"optimizer_steps":640,"epochs_completed":10,"training_seconds":158.38817469199967,"val_map50_95":0.7236498083277201,"roles":"tuning"},{"run_id":"journal-wildlife-MuSGD-s11-121fe76d19da","dataset":"wildlife","optimizer":"MuSGD","seed":11,"data_fraction":1.0,"noise_probability":0.0,"realized_noise_rate":0.0,"lr0":0.003,"train_images":1020,"images_seen":10240,"optimizer_steps":640,"epochs_completed":10,"training_seconds":169.30539370000042,"val_map50_95":0.8132552827839152,"roles":"tuning"},{"run_id":"journal-wildlife-MuSGD-s11-126d4ffc3001","dataset":"wildlife","optimizer":"MuSGD","seed":11,"data_fraction":0.2,"noise_probability":0.2,"realized_noise_rate":0.1875,"lr0":0.003,"train_images":204,"images_seen":10240,"optimizer_steps":640,"epochs_completed":10,"training_seconds":169.15288772199983,"val_map50_95":0.7351404674211354,"roles":"tuning"},{"run_id":"journal-wildlife-SGD-s11-99155deefa63","dataset":"wildlife","optimizer":"SGD","seed":11,"data_fraction":1.0,"noise_probability":0.0,"realized_noise_rate":0.0,"lr0":0.003,"train_images":1020,"images_seen":10240,"optimizer_steps":640,"epochs_completed":10,"training_seconds":157.47196234200055,"val_map50_95":0.8249458386297596,"roles":"tuning"},{"run_id":"journal-wildlife-SGD-s11-9d2e5b8a69d4","dataset":"wildlife","optimizer":"SGD","seed":11,"data_fraction":0.2,"noise_probability":0.2,"realized_noise_rate":0.1875,"lr0":0.003,"train_images":204,"images_seen":10240,"optimizer_steps":640,"epochs_completed":10,"training_seconds":157.25605108000036,"val_map50_95":0.7482405224469451,"roles":"tuning"},{"run_id":"journal-wildlife-AdamW-s11-ce1208ebdb62","dataset":"wildlife","optimizer":"AdamW","seed":11,"data_fraction":1.0,"noise_probability":0.0,"realized_noise_rate":0.0,"lr0":0.0003,"train_images":1020,"images_seen":10240,"optimizer_steps":640,"epochs_completed":10,"training_seconds":158.6950917439999,"val_map50_95":0.8176629004855659,"roles":"tuning"},{"run_id":"journal-wildlife-AdamW-s11-e8f1c2c8816c","dataset":"wildlife","optimizer":"AdamW","seed":11,"data_fraction":0.2,"noise_probability":0.2,"realized_noise_rate":0.1875,"lr0":0.0003,"train_images":204,"images_seen":10240,"optimizer_steps":640,"epochs_completed":10,"training_seconds":158.57215519700003,"val_map50_95":0.7380763803428776,"roles":"tuning"},{"run_id":"journal-wildlife-MuSGD-s11-27b96f06abaa","dataset":"wildlife","optimizer":"MuSGD","seed":11,"data_fraction":1.0,"noise_probability":0.0,"realized_noise_rate":0.0,"lr0":0.003,"train_images":1020,"images_seen":10240,"optimizer_steps":640,"epochs_completed":10,"training_seconds":169.9004064870005,"val_map50_95":0.8231134070584597,"roles":"tuning"},{"run_id":"journal-wildlife-MuSGD-s11-78db064150c1","dataset":"wildlife","optimizer":"MuSGD","seed":11,"data_fraction":0.2,"noise_probability":0.2,"realized_noise_rate":0.1875,"lr0":0.003,"train_images":204,"images_seen":10240,"optimizer_steps":640,"epochs_completed":10,"training_seconds":169.25997867099977,"val_map50_95":0.7345606141534515,"roles":"tuning"},{"run_id":"journal-wildlife-SGD-s11-3d2ed7f9c459","dataset":"wildlife","optimizer":"SGD","seed":11,"data_fraction":1.0,"noise_probability":0.0,"realized_noise_rate":0.0,"lr0":0.003,"train_images":1020,"images_seen":10240,"optimizer_steps":640,"epochs_completed":10,"training_seconds":157.85798799700024,"val_map50_95":0.8163789490269597,"roles":"tuning"},{"run_id":"journal-wildlife-SGD-s11-a0f457d8be6c","dataset":"wildlife","optimizer":"SGD","seed":11,"data_fraction":0.2,"noise_probability":0.2,"realized_noise_rate":0.1875,"lr0":0.003,"train_images":204,"images_seen":10240,"optimizer_steps":640,"epochs_completed":10,"training_seconds":157.69979561699984,"val_map50_95":0.7385803655406789,"roles":"tuning"},{"run_id":"journal-wildlife-AdamW-s11-004a91a00094","dataset":"wildlife","optimizer":"AdamW","seed":11,"data_fraction":1.0,"noise_probability":0.0,"realized_noise_rate":0.0,"lr0":0.001,"train_images":1020,"images_seen":10240,"optimizer_steps":640,"epochs_completed":10,"training_seconds":159.0934103740001,"val_map50_95":0.7629017907574606,"roles":"tuning"},{"run_id":"journal-wildlife-AdamW-s11-76ae91cf05dd","dataset":"wildlife","optimizer":"AdamW","seed":11,"data_fraction":0.2,"noise_probability":0.2,"realized_noise_rate":0.1875,"lr0":0.001,"train_images":204,"images_seen":10240,"optimizer_steps":640,"epochs_completed":10,"training_seconds":158.65484555799958,"val_map50_95":0.668431151807533,"roles":"tuning"},{"run_id":"journal-wildlife-MuSGD-s11-9c33db504b5e","dataset":"wildlife","optimizer":"MuSGD","seed":11,"data_fraction":1.0,"noise_probability":0.0,"realized_noise_rate":0.0,"lr0":0.01,"train_images":1020,"images_seen":10240,"optimizer_steps":640,"epochs_completed":10,"training_seconds":170.10021563699956,"val_map50_95":0.761591438798438,"roles":"tuning"},{"run_id":"journal-wildlife-MuSGD-s11-5a9bd7ed8d2b","dataset":"wildlife","optimizer":"MuSGD","seed":11,"data_fraction":0.2,"noise_probability":0.2,"realized_noise_rate":0.1875,"lr0":0.01,"train_images":204,"images_seen":10240,"optimizer_steps":640,"epochs_completed":10,"training_seconds":169.89684401399973,"val_map50_95":0.6501367970964302,"roles":"tuning"},{"run_id":"journal-wildlife-SGD-s11-1b7c385c5fe5","dataset":"wildlife","optimizer":"SGD","seed":11,"data_fraction":1.0,"noise_probability":0.0,"realized_noise_rate":0.0,"lr0":0.01,"train_images":1020,"images_seen":10240,"optimizer_steps":640,"epochs_completed":10,"training_seconds":158.05401196000003,"val_map50_95":0.7900610862811694,"roles":"tuning"},{"run_id":"journal-wildlife-SGD-s11-26511b0bc78b","dataset":"wildlife","optimizer":"SGD","seed":11,"data_fraction":0.2,"noise_probability":0.2,"realized_noise_rate":0.1875,"lr0":0.01,"train_images":204,"images_seen":10240,"optimizer_steps":640,"epochs_completed":10,"training_seconds":157.45506886299972,"val_map50_95":0.7068113107037295,"roles":"tuning"},{"run_id":"journal-wildlife-AdamW-s11-3bc524349c24","dataset":"wildlife","optimizer":"AdamW","seed":11,"data_fraction":1.0,"noise_probability":0.0,"realized_noise_rate":0.0,"lr0":0.001,"train_images":1020,"images_seen":10240,"optimizer_steps":640,"epochs_completed":10,"training_seconds":158.77904489699995,"val_map50_95":0.7725968080916482,"roles":"tuning"},{"run_id":"journal-wildlife-AdamW-s11-806ec3ef5317","dataset":"wildlife","optimizer":"AdamW","seed":11,"data_fraction":0.2,"noise_probability":0.2,"realized_noise_rate":0.1875,"lr0":0.001,"train_images":204,"images_seen":10240,"optimizer_steps":640,"epochs_completed":10,"training_seconds":158.41085229999953,"val_map50_95":0.6549326341020355,"roles":"tuning"},{"run_id":"journal-wildlife-MuSGD-s11-2b56dd85058e","dataset":"wildlife","optimizer":"MuSGD","seed":11,"data_fraction":1.0,"noise_probability":0.0,"realized_noise_rate":0.0,"lr0":0.01,"train_images":1020,"images_seen":10240,"optimizer_steps":640,"epochs_completed":10,"training_seconds":169.92614861600032,"val_map50_95":0.764835996310918,"roles":"tuning"},{"run_id":"journal-wildlife-MuSGD-s11-48a35f95c30d","dataset":"wildlife","optimizer":"MuSGD","seed":11,"data_fraction":0.2,"noise_probability":0.2,"realized_noise_rate":0.1875,"lr0":0.01,"train_images":204,"images_seen":10240,"optimizer_steps":640,"epochs_completed":10,"training_seconds":169.07149194499925,"val_map50_95":0.6381573021289725,"roles":"tuning"},{"run_id":"journal-wildlife-SGD-s11-a8010e579566","dataset":"wildlife","optimizer":"SGD","seed":11,"data_fraction":1.0,"noise_probability":0.0,"realized_noise_rate":0.0,"lr0":0.01,"train_images":1020,"images_seen":10240,"optimizer_steps":640,"epochs_completed":10,"training_seconds":157.5587880990006,"val_map50_95":0.773212863345101,"roles":"tuning"},{"run_id":"journal-wildlife-SGD-s11-76c686c5a86a","dataset":"wildlife","optimizer":"SGD","seed":11,"data_fraction":0.2,"noise_probability":0.2,"realized_noise_rate":0.1875,"lr0":0.01,"train_images":204,"images_seen":10240,"optimizer_steps":640,"epochs_completed":10,"training_seconds":157.32833406100144,"val_map50_95":0.6899489254700427,"roles":"tuning"},{"run_id":"journal-wildlife-AdamW-s11-6956aa092e9a","dataset":"wildlife","optimizer":"AdamW","seed":11,"data_fraction":1.0,"noise_probability":0.0,"realized_noise_rate":0.0,"lr0":0.003,"train_images":1020,"images_seen":10240,"optimizer_steps":640,"epochs_completed":10,"training_seconds":159.47889248799947,"val_map50_95":0.4491071638958896,"roles":"tuning"},{"run_id":"journal-wildlife-AdamW-s11-968b2e7d00d0","dataset":"wildlife","optimizer":"AdamW","seed":11,"data_fraction":0.2,"noise_probability":0.2,"realized_noise_rate":0.1875,"lr0":0.003,"train_images":204,"images_seen":10240,"optimizer_steps":640,"epochs_completed":10,"training_seconds":158.7745939560009,"val_map50_95":0.4269972951849838,"roles":"tuning"},{"run_id":"journal-wildlife-MuSGD-s11-e5712d62bf0c","dataset":"wildlife","optimizer":"MuSGD","seed":11,"data_fraction":1.0,"noise_probability":0.0,"realized_noise_rate":0.0,"lr0":0.03,"train_images":1020,"images_seen":10240,"optimizer_steps":640,"epochs_completed":10,"training_seconds":170.26203949400042,"val_map50_95":0.3281559123322362,"roles":"tuning"},{"run_id":"journal-wildlife-MuSGD-s11-74a8ab0fb5df","dataset":"wildlife","optimizer":"MuSGD","seed":11,"data_fraction":0.2,"noise_probability":0.2,"realized_noise_rate":0.1875,"lr0":0.03,"train_images":204,"images_seen":10240,"optimizer_steps":640,"epochs_completed":10,"training_seconds":170.4018004240006,"val_map50_95":0.19012396743197074,"roles":"tuning"},{"run_id":"journal-wildlife-SGD-s11-63f9f94c1f85","dataset":"wildlife","optimizer":"SGD","seed":11,"data_fraction":1.0,"noise_probability":0.0,"realized_noise_rate":0.0,"lr0":0.03,"train_images":1020,"images_seen":10240,"optimizer_steps":640,"epochs_completed":10,"training_seconds":157.47391178900034,"val_map50_95":0.5835719391552223,"roles":"tuning"},{"run_id":"journal-wildlife-SGD-s11-2a686e2aa8fb","dataset":"wildlife","optimizer":"SGD","seed":11,"data_fraction":0.2,"noise_probability":0.2,"realized_noise_rate":0.1875,"lr0":0.03,"train_images":204,"images_seen":10240,"optimizer_steps":640,"epochs_completed":10,"training_seconds":157.73940380599925,"val_map50_95":0.4115952150022421,"roles":"tuning"},{"run_id":"journal-wildlife-AdamW-s11-efb23d14ddbb","dataset":"wildlife","optimizer":"AdamW","seed":11,"data_fraction":1.0,"noise_probability":0.0,"realized_noise_rate":0.0,"lr0":0.003,"train_images":1020,"images_seen":10240,"optimizer_steps":640,"epochs_completed":10,"training_seconds":159.09248794299856,"val_map50_95":0.4864939648801128,"roles":"tuning"},{"run_id":"journal-wildlife-AdamW-s11-3d216439b8de","dataset":"wildlife","optimizer":"AdamW","seed":11,"data_fraction":0.2,"noise_probability":0.2,"realized_noise_rate":0.1875,"lr0":0.003,"train_images":204,"images_seen":10240,"optimizer_steps":640,"epochs_completed":10,"training_seconds":158.46474786300132,"val_map50_95":0.4306265830946477,"roles":"tuning"},{"run_id":"journal-wildlife-MuSGD-s11-0b2b6ec80397","dataset":"wildlife","optimizer":"MuSGD","seed":11,"data_fraction":1.0,"noise_probability":0.0,"realized_noise_rate":0.0,"lr0":0.03,"train_images":1020,"images_seen":10240,"optimizer_steps":640,"epochs_completed":10,"training_seconds":169.89593895100006,"val_map50_95":0.39796996295532894,"roles":"tuning"},{"run_id":"journal-wildlife-MuSGD-s11-22f01a54222c","dataset":"wildlife","optimizer":"MuSGD","seed":11,"data_fraction":0.2,"noise_probability":0.2,"realized_noise_rate":0.1875,"lr0":0.03,"train_images":204,"images_seen":10240,"optimizer_steps":640,"epochs_completed":10,"training_seconds":169.68434923399946,"val_map50_95":0.3094590092684485,"roles":"tuning"},{"run_id":"journal-wildlife-SGD-s11-8cd3cec78ff9","dataset":"wildlife","optimizer":"SGD","seed":11,"data_fraction":1.0,"noise_probability":0.0,"realized_noise_rate":0.0,"lr0":0.03,"train_images":1020,"images_seen":10240,"optimizer_steps":640,"epochs_completed":10,"training_seconds":158.29833196300024,"val_map50_95":0.5663015974636647,"roles":"tuning"},{"run_id":"journal-wildlife-SGD-s11-5dd00e831bae","dataset":"wildlife","optimizer":"SGD","seed":11,"data_fraction":0.2,"noise_probability":0.2,"realized_noise_rate":0.1875,"lr0":0.03,"train_images":204,"images_seen":10240,"optimizer_steps":640,"epochs_completed":10,"training_seconds":157.34705347099953,"val_map50_95":0.3683392113980551,"roles":"tuning"},{"run_id":"journal-wildlife-SGD-s101-065a344b0c31","dataset":"wildlife","optimizer":"SGD","seed":101,"data_fraction":0.2,"noise_probability":0.0,"realized_noise_rate":0.0,"lr0":0.003,"train_images":204,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":312.3632487240011,"val_map50_95":0.7766527689083249,"roles":"wildlife_confirmation_101"},{"run_id":"journal-wildlife-MuSGD-s101-b28c1f74adeb","dataset":"wildlife","optimizer":"MuSGD","seed":101,"data_fraction":0.2,"noise_probability":0.0,"realized_noise_rate":0.0,"lr0":0.003,"train_images":204,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":335.0382500320011,"val_map50_95":0.76898107273043,"roles":"wildlife_confirmation_101"},{"run_id":"journal-wildlife-AdamW-s101-52949b142cc1","dataset":"wildlife","optimizer":"AdamW","seed":101,"data_fraction":0.2,"noise_probability":0.0,"realized_noise_rate":0.0,"lr0":0.0003,"train_images":204,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":313.1010964980014,"val_map50_95":0.7799847787534946,"roles":"wildlife_confirmation_101"},{"run_id":"journal-wildlife-AdamW-s101-a54276b340b4","dataset":"wildlife","optimizer":"AdamW","seed":101,"data_fraction":0.2,"noise_probability":0.2,"realized_noise_rate":0.24210526315789474,"lr0":0.0003,"train_images":204,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":313.1499724719997,"val_map50_95":0.7344201047074046,"roles":"wildlife_confirmation_101"},{"run_id":"journal-wildlife-MuSGD-s101-0858f11dc137","dataset":"wildlife","optimizer":"MuSGD","seed":101,"data_fraction":0.2,"noise_probability":0.2,"realized_noise_rate":0.24210526315789474,"lr0":0.003,"train_images":204,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":335.14619871099967,"val_map50_95":0.7133315517845487,"roles":"wildlife_confirmation_101"},{"run_id":"journal-wildlife-SGD-s101-684fd74da7c4","dataset":"wildlife","optimizer":"SGD","seed":101,"data_fraction":0.2,"noise_probability":0.2,"realized_noise_rate":0.24210526315789474,"lr0":0.003,"train_images":204,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":311.1309761020002,"val_map50_95":0.7396286171193803,"roles":"wildlife_confirmation_101"},{"run_id":"journal-wildlife-SGD-s101-f68a925bf3b1","dataset":"wildlife","optimizer":"SGD","seed":101,"data_fraction":1.0,"noise_probability":0.0,"realized_noise_rate":0.0,"lr0":0.003,"train_images":1020,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":312.674492492999,"val_map50_95":0.8294324319084485,"roles":"wildlife_confirmation_101"},{"run_id":"journal-wildlife-MuSGD-s101-7edd8c369971","dataset":"wildlife","optimizer":"MuSGD","seed":101,"data_fraction":1.0,"noise_probability":0.0,"realized_noise_rate":0.0,"lr0":0.003,"train_images":1020,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":335.35277179399964,"val_map50_95":0.8192917754111912,"roles":"wildlife_confirmation_101"},{"run_id":"journal-wildlife-AdamW-s101-5ffef6486b7d","dataset":"wildlife","optimizer":"AdamW","seed":101,"data_fraction":1.0,"noise_probability":0.0,"realized_noise_rate":0.0,"lr0":0.0003,"train_images":1020,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":313.3519844729999,"val_map50_95":0.8249851882530577,"roles":"wildlife_confirmation_101"},{"run_id":"journal-wildlife-AdamW-s101-b28f15a74498","dataset":"wildlife","optimizer":"AdamW","seed":101,"data_fraction":1.0,"noise_probability":0.2,"realized_noise_rate":0.2154255319148936,"lr0":0.0003,"train_images":1020,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":315.0906748969992,"val_map50_95":0.7991070828730553,"roles":"wildlife_confirmation_101"},{"run_id":"journal-wildlife-MuSGD-s101-c111cdc335e2","dataset":"wildlife","optimizer":"MuSGD","seed":101,"data_fraction":1.0,"noise_probability":0.2,"realized_noise_rate":0.2154255319148936,"lr0":0.003,"train_images":1020,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":335.8127566300009,"val_map50_95":0.802824503085391,"roles":"wildlife_confirmation_101"},{"run_id":"journal-wildlife-SGD-s101-d005351952a4","dataset":"wildlife","optimizer":"SGD","seed":101,"data_fraction":1.0,"noise_probability":0.2,"realized_noise_rate":0.2154255319148936,"lr0":0.003,"train_images":1020,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":311.9174547879993,"val_map50_95":0.8093517206366853,"roles":"wildlife_confirmation_101"},{"run_id":"journal-wildlife-AdamW-s102-13195bcfeded","dataset":"wildlife","optimizer":"AdamW","seed":102,"data_fraction":0.2,"noise_probability":0.0,"realized_noise_rate":0.0,"lr0":0.0003,"train_images":204,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":313.61307021599896,"val_map50_95":0.7792508121126345,"roles":"wildlife_confirmation_102"},{"run_id":"journal-wildlife-MuSGD-s102-bb4799d08d80","dataset":"wildlife","optimizer":"MuSGD","seed":102,"data_fraction":0.2,"noise_probability":0.0,"realized_noise_rate":0.0,"lr0":0.003,"train_images":204,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":335.5452459070002,"val_map50_95":0.7604084984562588,"roles":"wildlife_confirmation_102"},{"run_id":"journal-wildlife-SGD-s102-b19436d4297b","dataset":"wildlife","optimizer":"SGD","seed":102,"data_fraction":0.2,"noise_probability":0.0,"realized_noise_rate":0.0,"lr0":0.003,"train_images":204,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":311.2967061649997,"val_map50_95":0.7800426819471891,"roles":"wildlife_confirmation_102"},{"run_id":"journal-wildlife-SGD-s102-5954d7538fa0","dataset":"wildlife","optimizer":"SGD","seed":102,"data_fraction":0.2,"noise_probability":0.2,"realized_noise_rate":0.18911917098445596,"lr0":0.003,"train_images":204,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":311.8122932739989,"val_map50_95":0.7261795989748115,"roles":"wildlife_confirmation_102"},{"run_id":"journal-wildlife-MuSGD-s102-b9600b8bed2c","dataset":"wildlife","optimizer":"MuSGD","seed":102,"data_fraction":0.2,"noise_probability":0.2,"realized_noise_rate":0.18911917098445596,"lr0":0.003,"train_images":204,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":334.2661198700007,"val_map50_95":0.7163716661707541,"roles":"wildlife_confirmation_102"},{"run_id":"journal-wildlife-AdamW-s102-859416b59dfb","dataset":"wildlife","optimizer":"AdamW","seed":102,"data_fraction":0.2,"noise_probability":0.2,"realized_noise_rate":0.18911917098445596,"lr0":0.0003,"train_images":204,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":313.71593327200026,"val_map50_95":0.7217724005324717,"roles":"wildlife_confirmation_102"},{"run_id":"journal-wildlife-AdamW-s102-c9afd78c1d9e","dataset":"wildlife","optimizer":"AdamW","seed":102,"data_fraction":1.0,"noise_probability":0.0,"realized_noise_rate":0.0,"lr0":0.0003,"train_images":1020,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":314.64342099900205,"val_map50_95":0.8239094378291612,"roles":"wildlife_confirmation_102"},{"run_id":"journal-wildlife-MuSGD-s102-e1f09d16d958","dataset":"wildlife","optimizer":"MuSGD","seed":102,"data_fraction":1.0,"noise_probability":0.0,"realized_noise_rate":0.0,"lr0":0.003,"train_images":1020,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":334.99847971599957,"val_map50_95":0.8224747916937669,"roles":"wildlife_confirmation_102"},{"run_id":"journal-wildlife-SGD-s102-6b6ffd8901ea","dataset":"wildlife","optimizer":"SGD","seed":102,"data_fraction":1.0,"noise_probability":0.0,"realized_noise_rate":0.0,"lr0":0.003,"train_images":1020,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":311.76692308400015,"val_map50_95":0.8395593811566286,"roles":"wildlife_confirmation_102"},{"run_id":"journal-wildlife-SGD-s102-9990295494d8","dataset":"wildlife","optimizer":"SGD","seed":102,"data_fraction":1.0,"noise_probability":0.2,"realized_noise_rate":0.19627659574468084,"lr0":0.003,"train_images":1020,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":311.5736617920011,"val_map50_95":0.8036737684890486,"roles":"wildlife_confirmation_102"},{"run_id":"journal-wildlife-MuSGD-s102-f535cf5f4d99","dataset":"wildlife","optimizer":"MuSGD","seed":102,"data_fraction":1.0,"noise_probability":0.2,"realized_noise_rate":0.19627659574468084,"lr0":0.003,"train_images":1020,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":334.7301851539996,"val_map50_95":0.7938724340328814,"roles":"wildlife_confirmation_102"},{"run_id":"journal-wildlife-AdamW-s102-9593c7cffb89","dataset":"wildlife","optimizer":"AdamW","seed":102,"data_fraction":1.0,"noise_probability":0.2,"realized_noise_rate":0.19627659574468084,"lr0":0.0003,"train_images":1020,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":313.4513470480015,"val_map50_95":0.7870889428643302,"roles":"wildlife_confirmation_102"},{"run_id":"journal-wildlife-SGD-s103-798239832b59","dataset":"wildlife","optimizer":"SGD","seed":103,"data_fraction":0.2,"noise_probability":0.0,"realized_noise_rate":0.0,"lr0":0.003,"train_images":204,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":311.61197685500156,"val_map50_95":0.7797652863427988,"roles":"wildlife_confirmation_103"},{"run_id":"journal-wildlife-MuSGD-s103-1a2ac4069efa","dataset":"wildlife","optimizer":"MuSGD","seed":103,"data_fraction":0.2,"noise_probability":0.0,"realized_noise_rate":0.0,"lr0":0.003,"train_images":204,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":335.53882535999946,"val_map50_95":0.7641709173160274,"roles":"wildlife_confirmation_103"},{"run_id":"journal-wildlife-AdamW-s103-60eb206ab1e7","dataset":"wildlife","optimizer":"AdamW","seed":103,"data_fraction":0.2,"noise_probability":0.0,"realized_noise_rate":0.0,"lr0":0.0003,"train_images":204,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":313.4318492759994,"val_map50_95":0.7556092842073505,"roles":"wildlife_confirmation_103"},{"run_id":"journal-wildlife-AdamW-s103-cd2499e1209d","dataset":"wildlife","optimizer":"AdamW","seed":103,"data_fraction":0.2,"noise_probability":0.2,"realized_noise_rate":0.16621983914209115,"lr0":0.0003,"train_images":204,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":312.9863359890005,"val_map50_95":0.731667428208471,"roles":"wildlife_confirmation_103"},{"run_id":"journal-wildlife-MuSGD-s103-14e598cfbbfd","dataset":"wildlife","optimizer":"MuSGD","seed":103,"data_fraction":0.2,"noise_probability":0.2,"realized_noise_rate":0.16621983914209115,"lr0":0.003,"train_images":204,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":335.4091797589972,"val_map50_95":0.717140751624813,"roles":"wildlife_confirmation_103"},{"run_id":"journal-wildlife-SGD-s103-5efcbf91c907","dataset":"wildlife","optimizer":"SGD","seed":103,"data_fraction":0.2,"noise_probability":0.2,"realized_noise_rate":0.16621983914209115,"lr0":0.003,"train_images":204,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":310.81635343000016,"val_map50_95":0.7222508950321733,"roles":"wildlife_confirmation_103"},{"run_id":"journal-wildlife-SGD-s103-7f317a10a6c4","dataset":"wildlife","optimizer":"SGD","seed":103,"data_fraction":1.0,"noise_probability":0.0,"realized_noise_rate":0.0,"lr0":0.003,"train_images":1020,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":312.15849536600217,"val_map50_95":0.8161824815619818,"roles":"wildlife_confirmation_103"},{"run_id":"journal-wildlife-MuSGD-s103-c6c92d9d7a43","dataset":"wildlife","optimizer":"MuSGD","seed":103,"data_fraction":1.0,"noise_probability":0.0,"realized_noise_rate":0.0,"lr0":0.003,"train_images":1020,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":335.0194770179987,"val_map50_95":0.8182426690485457,"roles":"wildlife_confirmation_103"},{"run_id":"journal-wildlife-AdamW-s103-f0cae4fccf5d","dataset":"wildlife","optimizer":"AdamW","seed":103,"data_fraction":1.0,"noise_probability":0.0,"realized_noise_rate":0.0,"lr0":0.0003,"train_images":1020,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":313.14646366099987,"val_map50_95":0.8163556237087836,"roles":"wildlife_confirmation_103"},{"run_id":"journal-wildlife-AdamW-s103-7dc2ab39ed62","dataset":"wildlife","optimizer":"AdamW","seed":103,"data_fraction":1.0,"noise_probability":0.2,"realized_noise_rate":0.20319148936170212,"lr0":0.0003,"train_images":1020,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":314.3460745119992,"val_map50_95":0.7921419747082368,"roles":"wildlife_confirmation_103"},{"run_id":"journal-wildlife-MuSGD-s103-4aa00f3e0081","dataset":"wildlife","optimizer":"MuSGD","seed":103,"data_fraction":1.0,"noise_probability":0.2,"realized_noise_rate":0.20319148936170212,"lr0":0.003,"train_images":1020,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":335.03361409600257,"val_map50_95":0.8001847472680463,"roles":"wildlife_confirmation_103"},{"run_id":"journal-wildlife-SGD-s103-89329bc5f74a","dataset":"wildlife","optimizer":"SGD","seed":103,"data_fraction":1.0,"noise_probability":0.2,"realized_noise_rate":0.20319148936170212,"lr0":0.003,"train_images":1020,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":311.58235832599894,"val_map50_95":0.794971658803455,"roles":"wildlife_confirmation_103"},{"run_id":"journal-voc2007-MuSGD-s101-e1e87dd429af","dataset":"voc2007","optimizer":"MuSGD","seed":101,"data_fraction":0.2,"noise_probability":0.0,"realized_noise_rate":0.0,"lr0":0.003,"train_images":501,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":377.71137492300113,"val_map50_95":0.4981167078270657,"roles":"voc2007_confirmation_101"},{"run_id":"journal-voc2007-AdamW-s101-db8b0bc4bdb6","dataset":"voc2007","optimizer":"AdamW","seed":101,"data_fraction":0.2,"noise_probability":0.0,"realized_noise_rate":0.0,"lr0":0.0003,"train_images":501,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":350.85303151900007,"val_map50_95":0.4776468094996794,"roles":"voc2007_confirmation_101"},{"run_id":"journal-voc2007-AdamW-s101-beca620c6fae","dataset":"voc2007","optimizer":"AdamW","seed":101,"data_fraction":0.2,"noise_probability":0.2,"realized_noise_rate":0.2144133412745682,"lr0":0.0003,"train_images":501,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":351.44179146999886,"val_map50_95":0.43306430738969726,"roles":"voc2007_confirmation_101"},{"run_id":"journal-voc2007-MuSGD-s101-32d0c431f3b2","dataset":"voc2007","optimizer":"MuSGD","seed":101,"data_fraction":0.2,"noise_probability":0.2,"realized_noise_rate":0.2144133412745682,"lr0":0.003,"train_images":501,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":371.22158058399873,"val_map50_95":0.45139104597495633,"roles":"voc2007_confirmation_101"},{"run_id":"journal-voc2007-MuSGD-s101-ec004384213a","dataset":"voc2007","optimizer":"MuSGD","seed":101,"data_fraction":1.0,"noise_probability":0.0,"realized_noise_rate":0.0,"lr0":0.003,"train_images":2501,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":374.88050417600243,"val_map50_95":0.5587531652361157,"roles":"voc2007_confirmation_101"},{"run_id":"journal-voc2007-AdamW-s101-f7f9ad8fc282","dataset":"voc2007","optimizer":"AdamW","seed":101,"data_fraction":1.0,"noise_probability":0.0,"realized_noise_rate":0.0,"lr0":0.0003,"train_images":2501,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":352.49232199400285,"val_map50_95":0.5528173070993662,"roles":"voc2007_confirmation_101"},{"run_id":"journal-voc2007-AdamW-s101-db0d00a476ce","dataset":"voc2007","optimizer":"AdamW","seed":101,"data_fraction":1.0,"noise_probability":0.2,"realized_noise_rate":0.20002549719530852,"lr0":0.0003,"train_images":2501,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":352.9282027110021,"val_map50_95":0.5310865332996615,"roles":"voc2007_confirmation_101"},{"run_id":"journal-voc2007-MuSGD-s101-e44491dd7acd","dataset":"voc2007","optimizer":"MuSGD","seed":101,"data_fraction":1.0,"noise_probability":0.2,"realized_noise_rate":0.20002549719530852,"lr0":0.003,"train_images":2501,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":372.7714328479997,"val_map50_95":0.5484946277803536,"roles":"voc2007_confirmation_101"},{"run_id":"journal-voc2007-AdamW-s102-85c47be03a6f","dataset":"voc2007","optimizer":"AdamW","seed":102,"data_fraction":0.2,"noise_probability":0.0,"realized_noise_rate":0.0,"lr0":0.0003,"train_images":501,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":352.2281241790006,"val_map50_95":0.47638267656591227,"roles":"voc2007_confirmation_102"},{"run_id":"journal-voc2007-MuSGD-s102-16f9bba198da","dataset":"voc2007","optimizer":"MuSGD","seed":102,"data_fraction":0.2,"noise_probability":0.0,"realized_noise_rate":0.0,"lr0":0.003,"train_images":501,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":372.0487217150003,"val_map50_95":0.4999345372075691,"roles":"voc2007_confirmation_102"},{"run_id":"journal-voc2007-MuSGD-s102-7ecf7a7942ec","dataset":"voc2007","optimizer":"MuSGD","seed":102,"data_fraction":0.2,"noise_probability":0.2,"realized_noise_rate":0.20550458715596331,"lr0":0.003,"train_images":501,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":371.6477334480005,"val_map50_95":0.4546455459411514,"roles":"voc2007_confirmation_102"},{"run_id":"journal-voc2007-AdamW-s102-da7f31d8979c","dataset":"voc2007","optimizer":"AdamW","seed":102,"data_fraction":0.2,"noise_probability":0.2,"realized_noise_rate":0.20550458715596331,"lr0":0.0003,"train_images":501,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":350.96718611599863,"val_map50_95":0.43800756331850693,"roles":"voc2007_confirmation_102"},{"run_id":"journal-voc2007-AdamW-s102-fcccfffb5af4","dataset":"voc2007","optimizer":"AdamW","seed":102,"data_fraction":1.0,"noise_probability":0.0,"realized_noise_rate":0.0,"lr0":0.0003,"train_images":2501,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":354.0955436740005,"val_map50_95":0.5526249035747693,"roles":"voc2007_confirmation_102"},{"run_id":"journal-voc2007-MuSGD-s102-69fefb375fa2","dataset":"voc2007","optimizer":"MuSGD","seed":102,"data_fraction":1.0,"noise_probability":0.0,"realized_noise_rate":0.0,"lr0":0.003,"train_images":2501,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":372.366876398999,"val_map50_95":0.5543662962675667,"roles":"voc2007_confirmation_102"},{"run_id":"journal-voc2007-MuSGD-s102-028ab7c224d5","dataset":"voc2007","optimizer":"MuSGD","seed":102,"data_fraction":1.0,"noise_probability":0.2,"realized_noise_rate":0.20805711371749108,"lr0":0.003,"train_images":2501,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":374.4977631319998,"val_map50_95":0.5528619797998041,"roles":"voc2007_confirmation_102"},{"run_id":"journal-voc2007-AdamW-s102-bb6b06deb732","dataset":"voc2007","optimizer":"AdamW","seed":102,"data_fraction":1.0,"noise_probability":0.2,"realized_noise_rate":0.20805711371749108,"lr0":0.0003,"train_images":2501,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":351.68938790299944,"val_map50_95":0.5264523427336026,"roles":"voc2007_confirmation_102"},{"run_id":"journal-voc2007-MuSGD-s103-78d477f03af4","dataset":"voc2007","optimizer":"MuSGD","seed":103,"data_fraction":0.2,"noise_probability":0.0,"realized_noise_rate":0.0,"lr0":0.003,"train_images":501,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":372.5055078700025,"val_map50_95":0.4878050271872897,"roles":"voc2007_confirmation_103"},{"run_id":"journal-voc2007-AdamW-s103-c015562acecd","dataset":"voc2007","optimizer":"AdamW","seed":103,"data_fraction":0.2,"noise_probability":0.0,"realized_noise_rate":0.0,"lr0":0.0003,"train_images":501,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":350.99778525700094,"val_map50_95":0.47827188924499275,"roles":"voc2007_confirmation_103"},{"run_id":"journal-voc2007-AdamW-s103-bff6632acca4","dataset":"voc2007","optimizer":"AdamW","seed":103,"data_fraction":0.2,"noise_probability":0.2,"realized_noise_rate":0.20174890693316677,"lr0":0.0003,"train_images":501,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":351.6803212390005,"val_map50_95":0.44122790120307187,"roles":"voc2007_confirmation_103"},{"run_id":"journal-voc2007-MuSGD-s103-860772e40426","dataset":"voc2007","optimizer":"MuSGD","seed":103,"data_fraction":0.2,"noise_probability":0.2,"realized_noise_rate":0.20174890693316677,"lr0":0.003,"train_images":501,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":371.9053644100022,"val_map50_95":0.4506866969876656,"roles":"voc2007_confirmation_103"},{"run_id":"journal-voc2007-MuSGD-s103-ad4afc7208e9","dataset":"voc2007","optimizer":"MuSGD","seed":103,"data_fraction":1.0,"noise_probability":0.0,"realized_noise_rate":0.0,"lr0":0.003,"train_images":2501,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":374.7128848270004,"val_map50_95":0.5558371494519538,"roles":"voc2007_confirmation_103"},{"run_id":"journal-voc2007-AdamW-s103-511958a75331","dataset":"voc2007","optimizer":"AdamW","seed":103,"data_fraction":1.0,"noise_probability":0.0,"realized_noise_rate":0.0,"lr0":0.0003,"train_images":2501,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":351.98643200600054,"val_map50_95":0.541182903008148,"roles":"voc2007_confirmation_103"},{"run_id":"journal-voc2007-AdamW-s103-87d0e9fa0835","dataset":"voc2007","optimizer":"AdamW","seed":103,"data_fraction":1.0,"noise_probability":0.2,"realized_noise_rate":0.2001529831718511,"lr0":0.0003,"train_images":2501,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":353.2473786570008,"val_map50_95":0.5320368005551578,"roles":"voc2007_confirmation_103"},{"run_id":"journal-voc2007-MuSGD-s103-f679de367434","dataset":"voc2007","optimizer":"MuSGD","seed":103,"data_fraction":1.0,"noise_probability":0.2,"realized_noise_rate":0.2001529831718511,"lr0":0.003,"train_images":2501,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":371.7135870409984,"val_map50_95":0.5373828486480475,"roles":"voc2007_confirmation_103"},{"run_id":"journal-wildlife-MuSGD-s101-f062c11789ac","dataset":"wildlife","optimizer":"MuSGD","seed":101,"data_fraction":0.2,"noise_probability":0.2,"realized_noise_rate":0.24210526315789474,"lr0":0.003,"train_images":204,"images_seen":61440,"optimizer_steps":3840,"epochs_completed":60,"training_seconds":1002.1566007750007,"val_map50_95":0.70682096362994,"roles":"long_schedule_transfer"},{"run_id":"journal-wildlife-AdamW-s101-eddb8838f7f0","dataset":"wildlife","optimizer":"AdamW","seed":101,"data_fraction":0.2,"noise_probability":0.2,"realized_noise_rate":0.24210526315789474,"lr0":0.0003,"train_images":204,"images_seen":61440,"optimizer_steps":3840,"epochs_completed":60,"training_seconds":952.0986191920019,"val_map50_95":0.7132831486130806,"roles":"long_schedule_transfer"},{"run_id":"journal-wildlife-AdamW-s101-15c50aace9eb","dataset":"wildlife","optimizer":"AdamW","seed":101,"data_fraction":1.0,"noise_probability":0.2,"realized_noise_rate":0.2154255319148936,"lr0":0.0003,"train_images":1020,"images_seen":61440,"optimizer_steps":3840,"epochs_completed":60,"training_seconds":954.2513163679978,"val_map50_95":0.7640210004521379,"roles":"long_schedule_transfer"},{"run_id":"journal-wildlife-MuSGD-s101-2996e0d09054","dataset":"wildlife","optimizer":"MuSGD","seed":101,"data_fraction":1.0,"noise_probability":0.2,"realized_noise_rate":0.2154255319148936,"lr0":0.003,"train_images":1020,"images_seen":61440,"optimizer_steps":3840,"epochs_completed":60,"training_seconds":1025.6064490799981,"val_map50_95":0.7822496732393366,"roles":"long_schedule_transfer"},{"run_id":"journal-wildlife-MuSGD-s101-0927d132adca","dataset":"wildlife","optimizer":"MuSGD","seed":101,"data_fraction":0.2,"noise_probability":0.0,"realized_noise_rate":0.0,"lr0":0.003,"train_images":204,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":342.4476039700021,"val_map50_95":0.7845514982360052,"roles":"head_lr_1x_ablation"},{"run_id":"journal-wildlife-MuSGD-s101-245e2643f413","dataset":"wildlife","optimizer":"MuSGD","seed":101,"data_fraction":0.2,"noise_probability":0.2,"realized_noise_rate":0.24210526315789474,"lr0":0.003,"train_images":204,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":342.17450693000137,"val_map50_95":0.727470413542828,"roles":"head_lr_1x_ablation"},{"run_id":"journal-wildlife-MuSGD-s101-457c8c813c0a","dataset":"wildlife","optimizer":"MuSGD","seed":101,"data_fraction":1.0,"noise_probability":0.0,"realized_noise_rate":0.0,"lr0":0.003,"train_images":1020,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":343.43700277600146,"val_map50_95":0.8201195096455056,"roles":"head_lr_1x_ablation"},{"run_id":"journal-wildlife-MuSGD-s101-bbe02cded838","dataset":"wildlife","optimizer":"MuSGD","seed":101,"data_fraction":1.0,"noise_probability":0.2,"realized_noise_rate":0.2154255319148936,"lr0":0.003,"train_images":1020,"images_seen":20480,"optimizer_steps":1280,"epochs_completed":20,"training_seconds":346.3767896969948,"val_map50_95":0.8023946129670708,"roles":"head_lr_1x_ablation"}],"provenance":{"original_version":1,"script_version_id":361947826,"url":"https://www.kaggle.com/code/myudak/muon-yolo-journal?scriptVersionId=361947826","file_sha256":{"campaign_summary.json":"0011f17e58733a925915eb282c72252d92453859f002d4f5417d1b70b826639f","results.json":"f9fab2e14fc0045f11c6ca491b16fc9b78ece0998b5d93485a3cd6e7cd4edf91","final_test_results.json":"ebd65b5fea5f266eef46b8fb283766600366e5e52a5abb3eabde418d2724547f","frozen_final_test_plan.json":"793ace1d384c2717638cea466c4023d5da0156a8b1af76b572b74751e25b839c","selected_hyperparameters.json":"c334c8b2bdd0e2f97976b1f1709b03d3283a9f77ec9190a96cfbeaf85b90ad86"},"independently_checked_summaries":70,"author":"Muchammad Yuda Tri Ananda","email":"yuda@myudak.com"}}'
assert hashlib.sha256(evidence_json.encode()).hexdigest()=='3d6f68dc188ff778d4814c808451535b10df6b06aee5732d682a9227efb0a658'
snapshot=json.loads(evidence_json)
"""Pure presentation of archived numeric evidence; contains no training code."""
import csv
import html
import io
import json
import math
import statistics
from pathlib import Path

import plotly.graph_objects as go
import plotly.io as pio
from IPython.display import HTML, IFrame, display

COLORS={'AdamW':'#64748b','MuSGD':'#2563eb','SGD':'#059669'}
NAMES={'wildlife':'African Wildlife','voc2007':'VOC2007-derived'}
SPLITS={'frozen_test':'Frozen test','validation':'Validation'}
CONDITIONS=[(0.2,0.0),(0.2,0.2),(1.0,0.0),(1.0,0.2)]
CONFIG={'displaylogo':False,'responsive':True,'toImageButtonOptions':{'format':'svg'}}
FIGURES={}
pio.templates.default='plotly_white'


def condition(f,q):
    return f'{f:.0%} images / {q:.0%} missing boxes'


def style(fig,title,ytitle=None,height=520):
    fig.update_layout(title={'text':title,'x':0.01,'font':{'size':21}},height=height,
        font={'family':'Arial, sans-serif','size':13,'color':'#24334b'},
        margin={'l':165,'r':35,'t':135,'b':75},paper_bgcolor='white',plot_bgcolor='white',
        legend={'orientation':'h','y':-0.18,'x':0},hoverlabel={'bgcolor':'white'},
        xaxis={'gridcolor':'#e8edf4','zerolinecolor':'#94a3b8'},
        yaxis={'gridcolor':'#e8edf4','title':ytitle})
    return fig


def menu(buttons,x=0):
    return {'buttons':buttons,'direction':'down','x':x,'y':1.20,'xanchor':'left',
        'yanchor':'top','showactive':True,'bgcolor':'#f1f5f9','bordercolor':'#cbd5e1'}


def show(name,fig):
    FIGURES[name]=fig
    # Kaggle's RequireJS can interfere with the connected Plotly renderer.
    # Isolate each chart in its own HTML document with a shared local JS bundle.
    on_kaggle=Path('/kaggle/working').exists()
    output=Path('/kaggle/working') if on_kaggle else Path('output/kaggle_report')
    output.mkdir(parents=True,exist_ok=True)
    fig.write_html(output/(name+'.html'),include_plotlyjs='directory',full_html=True,
        config=CONFIG,div_id=name)
    src=name+'.html' if on_kaggle else '../../output/kaggle_report/'+name+'.html'
    display(IFrame(src=src,width='100%',height=fig.layout.height+24))


def effects(snapshot):
    fig=go.Figure();groups=[]
    for split in ('frozen_test','validation'):
        for dataset in NAMES:
            indices=[]
            for optimizer,offset in [('MuSGD',-0.10),('SGD',0.10)]:
                records=[r for r in snapshot['summary'][split][dataset]['paired_optimizer_contrasts']
                         if r['contrast']==f'{optimizer} minus AdamW']
                if not records:continue
                y=[CONDITIONS.index((r['fraction'],r['missing_probability']))+offset for r in records]
                indices.append(len(fig.data))
                fig.add_trace(go.Scatter(x=[r['mean'] for r in records],y=y,mode='markers',
                    name=f'{optimizer} minus AdamW',marker={'size':12,'color':COLORS[optimizer],
                    'symbol':'circle' if optimizer=='MuSGD' else 'diamond'},visible=False,
                    customdata=[[condition(r['fraction'],r['missing_probability']),*r['ci95']] for r in records],
                    error_x={'type':'data','array':[r['ci95'][1]-r['mean'] for r in records],
                    'arrayminus':[r['mean']-r['ci95'][0] for r in records],'color':COLORS[optimizer],'thickness':2},
                    hovertemplate='%{customdata[0]}<br>Mean: %{x:+.3f} AP points<br>Approx. 95% interval: [%{customdata[1]:.3f}, %{customdata[2]:.3f}]<extra>%{fullData.name}</extra>'))
                indices.append(len(fig.data))
                fig.add_trace(go.Scatter(x=[v for r in records for v in r['values']],
                    y=[yi for yi in y for _ in range(3)],mode='markers',showlegend=False,visible=False,
                    marker={'size':6,'opacity':0.4,'color':COLORS[optimizer]},
                    customdata=[[b,condition(r['fraction'],r['missing_probability'])] for r in records for b in (101,102,103)],
                    hovertemplate='%{customdata[1]}<br>Joint block %{customdata[0]}: %{x:+.3f} AP points<extra>Paired block difference</extra>'))
            groups.append((f'{NAMES[dataset]} · {SPLITS[split]}',indices))
    for i in groups[0][1]:fig.data[i].visible=True
    buttons=[{'label':label,'method':'update','args':[{'visible':[i in indices for i in range(len(fig.data))]},
        {'title.text':f'Recipe differences · {label}'}]} for label,indices in groups]
    style(fig,'Recipe differences · '+groups[0][0],height=560)
    fig.update_layout(updatemenus=[menu(buttons)],xaxis_title='mAP50–95 difference (AP points); positive favors the named recipe',
        yaxis={'tickvals':list(range(4)),'ticktext':[condition(*c) for c in CONDITIONS],'autorange':'reversed'})
    fig.add_vline(x=0,line_dash='dash',line_color='#475569')
    return fig


def scores(snapshot):
    fig=go.Figure();groups=[]
    for split in ('frozen_test','validation'):
        for dataset in NAMES:
            indices=[]
            for optimizer in COLORS:
                rows=[r for r in snapshot['summary'][split][dataset]['conditions'] if r['optimizer']==optimizer]
                if not rows:continue
                indices.append(len(fig.data))
                fig.add_trace(go.Bar(name=optimizer,x=[f"{r['fraction']:.0%} images<br>{r['missing_probability']:.0%} missing boxes" for r in rows],
                    y=[r['mean'] for r in rows],marker_color=COLORS[optimizer],visible=False,
                    error_y={'type':'data','array':[r['ci95'][1]-r['mean'] for r in rows],
                        'arrayminus':[r['mean']-r['ci95'][0] for r in rows]},
                    text=[f"{r['mean']:.2f}" for r in rows],textposition='outside',
                    hovertemplate='%{x}<br>Mean mAP50–95: %{y:.3f} AP<extra>%{fullData.name}</extra>'))
            groups.append((f'{NAMES[dataset]} · {SPLITS[split]}',indices))
    for i in groups[0][1]:fig.data[i].visible=True
    buttons=[{'label':label,'method':'update','args':[{'visible':[i in indices for i in range(len(fig.data))]},
        {'title.text':'Absolute scores · '+label}]} for label,indices in groups]
    style(fig,'Absolute scores · '+groups[0][0],'mAP50–95 × 100 (AP)',height=565)
    fig.update_layout(barmode='group',margin_l=65,updatemenus=[menu(buttons)],yaxis_range=[0,100],xaxis_tickangle=0)
    return fig


def interaction(snapshot):
    fig=go.Figure();groups=[]
    for split in ('frozen_test','validation'):
        indices=[];labels=[];records=[]
        for dataset in NAMES:
            for row in snapshot['summary'][split][dataset]['three_way_interactions']:
                labels.append(NAMES[dataset]+'<br>'+row['contrast'].split(':')[0].replace(' minus ',' − '));records.append(row)
        indices.append(len(fig.data))
        fig.add_trace(go.Scatter(x=[r['mean'] for r in records],y=labels,mode='markers',visible=False,
            marker={'size':12,'color':'#7c3aed'},name='Three-way interaction',
            error_x={'type':'data','array':[r['ci95'][1]-r['mean'] for r in records],
                'arrayminus':[r['mean']-r['ci95'][0] for r in records]},
            customdata=[r['ci95'] for r in records],
            hovertemplate='%{y}<br>Interaction: %{x:+.3f} AP points<br>Approx. 95% interval: [%{customdata[0]:.3f}, %{customdata[1]:.3f}]<extra></extra>'))
        groups.append((SPLITS[split],indices))
    fig.data[0].visible=True
    buttons=[{'label':label,'method':'update','args':[{'visible':[i in indices for i in range(len(fig.data))]},
        {'title.text':'Scarcity × missingness interaction · '+label}]} for label,indices in groups]
    style(fig,'Scarcity × missingness interaction · Frozen test',height=440)
    fig.update_layout(margin_l=175,updatemenus=[menu(buttons)],showlegend=False,
        xaxis_title='Difference of paired recipe differences (AP points)',yaxis_autorange='reversed')
    fig.add_vline(x=0,line_dash='dash',line_color='#475569')
    return fig


def search(snapshot):
    fig=go.Figure();groups=[]
    for optimizer in COLORS:
        indices=[]
        rows=snapshot['selection']['diagnostics'][optimizer]['ranked']
        for decay in (0.0,0.0005):
            selected=sorted([r for r in rows if r['hyperparameters']['weight_decay']==decay],key=lambda r:r['hyperparameters']['lr0'])
            indices.append(len(fig.data))
            fig.add_trace(go.Scatter(x=[r['hyperparameters']['lr0'] for r in selected],
                y=[100*r['mean_anchor_val_map'] for r in selected],mode='lines+markers',visible=False,
                name=f'Weight decay {decay:g}',line={'color':COLORS[optimizer],'dash':'solid' if decay else 'dot'},
                marker={'size':10,'symbol':'circle' if decay else 'diamond'},
                hovertemplate='Base learning rate: %{x:g}<br>Mean anchor validation AP: %{y:.3f}<extra>%{fullData.name}</extra>'))
        best=rows[0];indices.append(len(fig.data))
        fig.add_trace(go.Scatter(x=[best['hyperparameters']['lr0']],y=[100*best['mean_anchor_val_map']],
            mode='markers',name='Selected candidate',visible=False,marker={'size':19,'symbol':'star','color':'#e59b20'},
            hovertemplate='Selected at the lower LR boundary<br>LR %{x:g} · validation AP %{y:.3f}<extra></extra>'))
        groups.append((optimizer,indices))
    for i in groups[0][1]:fig.data[i].visible=True
    style(fig,'Validation search · AdamW','Mean two-anchor validation mAP50–95 × 100',height=515)
    fig.update_layout(margin_l=80,xaxis_type='log',xaxis_title='Base learning rate (log scale)',yaxis_range=[0,100],
        updatemenus=[menu([{'label':label,'method':'update','args':[{'visible':[i in ids for i in range(len(fig.data))]},
            {'title.text':'Validation search · '+label}]} for label,ids in groups])])
    return fig


def curves(snapshot):
    fig=go.Figure();groups=[];ap=[];loss=[]
    for dataset in NAMES:
        for fraction,missing in CONDITIONS:
            indices=[]
            records=[r for r in snapshot['histories'] if r['dataset']==dataset and r['fraction']==fraction and r['missing']==missing]
            for row in sorted(records,key=lambda r:(r['optimizer'],r['block'])):
                indices.append(len(fig.data));ap.append(row['ap']);loss.append(row['box_loss'])
                fig.add_trace(go.Scatter(x=row['exposure'],y=row['ap'],mode='lines',visible=False,
                    name=f"{row['optimizer']} · block {row['block']}",
                    line={'color':COLORS[row['optimizer']],'dash':{101:'solid',102:'dot',103:'dash'}[row['block']],'width':2},
                    hovertemplate='Presented images: %{x:,}<br>Value: %{y:.3f}<extra>%{fullData.name}</extra>'))
            groups.append((NAMES[dataset]+' · '+condition(fraction,missing),indices))
    for i in groups[0][1]:fig.data[i].visible=True
    style(fig,'Learning trajectories','Clean validation mAP50–95 × 100',height=610)
    fig.update_layout(margin_l=80,xaxis_title='Actual image presentations (not unique images)',legend_y=-0.20,
        updatemenus=[menu([{'label':label,'method':'update','args':[{'visible':[i in ids for i in range(len(fig.data))]},
            {'title.text':'Learning trajectories'}]} for label,ids in groups]),
            menu([{'label':'Validation AP','method':'update','args':[{'y':ap},{'yaxis.title.text':'Clean validation mAP50–95 × 100'}]},
                  {'label':'Training box loss','method':'update','args':[{'y':loss},{'yaxis.title.text':'Training box loss'}]}],x=0.72)])
    return fig


def table(rows):
    columns=list(rows[0])
    return '<div style="overflow-x:auto"><table style="border-collapse:collapse;width:100%"><thead><tr>'+''.join(
        '<th style="text-align:left;padding:9px;border-bottom:2px solid #cbd5e1">'+html.escape(k)+'</th>' for k in columns
    )+'</tr></thead><tbody>'+''.join('<tr>'+''.join('<td style="padding:9px;border-bottom:1px solid #e2e8f0">'+html.escape(str(r[k]))+'</td>' for k in columns)+'</tr>' for r in rows)+'</tbody></table></div>'


def exact_table(snapshot):
    rows=[]
    for dataset in NAMES:
        group=snapshot['summary']['frozen_test'][dataset]
        for f,q in CONDITIONS:
            values={r['optimizer']:r['mean'] for r in group['conditions'] if (r['fraction'],r['missing_probability'])==(f,q)}
            difference=next(r for r in group['paired_optimizer_contrasts'] if r['contrast']=='MuSGD minus AdamW' and (r['fraction'],r['missing_probability'])==(f,q))
            rows.append({'Dataset':NAMES[dataset],'Images':f'{f:.0%}','Missing':f'{q:.0%}',
                'AdamW':f"{values['AdamW']:.2f}",'MuSGD':f"{values['MuSGD']:.2f}",
                'SGD':f"{values['SGD']:.2f}" if 'SGD' in values else 'Not evaluated',
                'MuSGD − AdamW':f"{difference['mean']:+.2f}",
                'Approx. 95% paired interval':f"[{difference['ci95'][0]:.2f}, {difference['ci95'][1]:.2f}]"})
    return table(rows)


def save_outputs(snapshot,sections):
    output=Path('/kaggle/working') if Path('/kaggle/working').exists() else Path('output/kaggle_report')
    output.mkdir(parents=True,exist_ok=True)
    (output/'reviewed_evidence.json').write_text(json.dumps(snapshot,indent=2),encoding='utf-8')
    with (output/'trial_metrics.csv').open('w',newline='',encoding='utf-8') as stream:
        writer=csv.DictWriter(stream,fieldnames=list(snapshot['trials'][0]));writer.writeheader();writer.writerows(snapshot['trials'])
    parts=['<!doctype html><html lang="en"><meta charset="utf-8"><meta name="viewport" content="width=device-width,initial-scale=1"><title>MuSGD × YOLO26 — Results Explorer</title>',
        '<style>body{font:16px/1.65 system-ui,sans-serif;color:#24334b;margin:0;background:#f1f5f9}main{max-width:1120px;margin:auto;padding:48px 30px;background:white}h1{font-size:38px;line-height:1.2}h2{margin-top:54px}a{color:#2563eb}.note{padding:20px;background:#eff6ff;border-left:4px solid #2563eb}.plotly-graph-div{max-width:100%}td,th{font-size:14px}code{font-size:13px;overflow-wrap:anywhere}</style><main>']
    first=True
    for heading,body,chart in sections:
        parts.append(f'<h2>{heading}</h2>' if heading else '')
        parts.append(body)
        if chart:
            parts.append(FIGURES[chart].to_html(full_html=False,include_plotlyjs=True if first else False,config=CONFIG,div_id=chart))
            first=False
    parts.append('</main></html>')
    (output/'muon_yolo_results_explorer.html').write_text('\n'.join(parts),encoding='utf-8')
    (output/'figure_specs.json').write_text(json.dumps({k:json.loads(f.to_json()) for k,f in FIGURES.items()}),encoding='utf-8')
    return output

show('paired-effects',effects(snapshot))
```

## 2. What was actually held constant?

| Design choice | What it means |
|:--|:--|
| Model | Same pinned pretrained YOLO26n checkpoint; Ultralytics 8.4.187 |
| Image availability | 20% or 100% of the training-image pool |

| Selection | Six candidates per optimizer, two wildlife validation anchors, ten exposure blocks; no test-driven tuning |
| Transfer | Wildlife-selected AdamW/MuSGD settings transferred to VOC-derived data without VOC tuning |
| Evaluation | Clean validation; one final evaluation of each of 60 frozen validation-selected checkpoints |

## 7. Controls, sensitivity and interpretation

**Exposure controls.** The earlier pilot's ranking reversal was absent even in the new variable-exposure bridge. On scarce wildlife images, its MuSGD–AdamW validation differences were −0.16 AP (clean) and −2.16 AP (missing boxes); the fixed-exposure controls gave −2.84 and −3.12. Source membership, cyclic sampling and batch rounding also changed, so the phase difference cannot be attributed solely to update matching.

**Longer schedules.** Four one-block 60-exposure-block transfer checks lowered best-checkpoint validation AP by 0.65–3.51 points relative to their corresponding 20-block runs. The horizon and schedule changed together; this does not identify an optimal horizon.

**Head learning rate.** Replacing MuSGD's native 3× selected-head LR with 1× changed block-101 validation AP by +1.56 / +1.41 / +0.08 / −0.04 across scarce-clean / scarce-missing / full-clean / full-missing conditions. This is a sensitivity check, not an independently retuned test comparison.

**What to take away.** This is a conditional comparison of existing library recipes. It provides measured, paired evidence under controlled exposure, while leaving the primary interaction unresolved. It does not establish a universally superior optimizer or isolate the orthogonalization mechanism.
